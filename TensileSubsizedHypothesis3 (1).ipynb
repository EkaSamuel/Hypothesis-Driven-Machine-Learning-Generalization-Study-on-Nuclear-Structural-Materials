4. Scaling
After unit conversion, scale all features together:

StandardScaler (mean=0, std=1) for models like linear regression, neural nets.

MinMaxScaler (0–1) for distance-based models like k-NN.

This prevents large features from dominating due to their magnitude.

5. Handling Missing ppm Values
If an element is below detection limit, ppm may be recorded as blank or zero:

If it’s truly zero, keep it as 0.

If it’s missing due to not being measured, impute (median is safer than mean for skewed data).

Keep in mind: imputation can introduce bias if the missingness is systematic.

6. Consider Feature Engineering
Ratios of ppm elements can be more predictive than absolute amounts for some metallurgy problems (e.g., Cr/Ni ratio).

Grouping: Combine certain ppm elements into “trace impurity index”.

✅ Rule of Thumb:
For machine learning, always bring ppm into the same scale and unit as other composition features before modeling. Otherwise, the algorithm might think that 500 ppm is "huge" compared to 0.5 wt.%, when in reality they’re almost the same magnitude in physical terms.

sequence of data manipulation is very important because you want to avoid data leakage or a situation where data modification afffects the integrity of other data in subsequent manipulations.

its best to start from the least complex, gradually manipulate the dataset, and ensure each action was carried out without affecting the integrity of the dataset.

```python
# Handling values such as <40 ppm, 150-400ppm, <0.6 & applying to all chemical composition columns



import pandas as pd
import numpy as np
import re

def parse_chem_value(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # Handle ppm ranges like "150-400ppm"
    range_match = re.match(r'(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        ppm_avg = (low + high) / 2
        return ppm_avg / 10000  # ppm to wt.%
    
    # Handle less-than values e.g. "<40 ppm" or "<0.6"
    lt_match = re.match(r'<\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if lt_match:
        num = float(lt_match.group(1))
        if "ppm" in val_str:
            return num / 10000  # ppm to wt.%
        else:
            return num  # already wt.%
    
    # Handle single ppm values e.g. "200 ppm"
    ppm_match = re.match(r'(\d+(?:\.\d+)?)(?:\s*ppm)', val_str)
    if ppm_match:
        num = float(ppm_match.group(1))
        return num / 10000  # ppm to wt.%
    
    # Handle plain numbers (already wt.%)
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# Columns to modify
cols_to_modify = [
    'C (wt.%)', 'Si (wt.%)', 'Mn (wt.%)', 'P (wt.%)', 'S (wt.%)',
    'Ni (wt.%)', 'Cr (wt.%)', 'Mo (wt.%)', 'Al (wt.%)', 'N (wt.%)',
    'Ti (wt.%)', 'Fe (wt.%)', 'Nb (wt.%)', 'B (wt.%)', 'Cu (wt.%)',
    'V (wt.%)', 'Mg (wt.%)', 'Y (wt.%)', 'Zr (wt.%)', 'Zn (wt.%)',
    'W (wt.%)', 'Sn (wt.%)', 'Co (wt.%)', 'O (wt.,%)', 'Ta (wt.%)',
    'H (wt.%)'
]

# Apply parsing function to each relevant column
for col in cols_to_modify:
    if col in df.columns:
        df[col] = df[col].apply(parse_chem_value)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfYS_clean, dfYS_outliers
    
    # Use df_remaining if it exists, else default to dfYS
    try:
        data_to_process = df_remaining if 'df_remaining' in globals() and df_remaining is not None else dfYS
    except NameError:
        data_to_process = dfYS

    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    dfYS_clean, dfYS_outliers = remover.fit_transform(data_to_process)

    with output_area:
        clear_output(wait=True)
        print(f"Original shape: {data_to_process.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Outliers shape: {dfYS_outliers.shape}")

        if show_full_checkbox.value:
            print("\n✅ Cleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_clean.to_html() + '</div>'))
            print("\n🚩 Outliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned DataFrame (first 5 rows):")
            display(HTML(dfYS_clean.head().to_html()))
            print("\n🚩 Outliers (first 5 rows):")
            display(HTML(dfYS_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

#### Preprocessing Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score

# Columns to aggregate by
group_cols = ['Chemistry & Alloy Family', 'Microstructure Class', 'Nuclear Application']

if df_holdout is not None and best_pipeline is not None:
    target_column = 'Yield Strength (MPa)'
    
    # Predict on holdout
    X_holdout = df_holdout.drop(columns=[target_column])
    y_holdout = df_holdout[target_column]
    y_holdout_pred = best_pipeline.predict(X_holdout)
    
    # Calculate overall R² score
    overall_r2 = r2_score(y_holdout, y_holdout_pred)
    print(f"📊 Overall R² Score on Holdout Set: {overall_r2:.4f}\n")
    
    # Combine results
    df_results = df_holdout.copy()
    df_results['Predicted'] = y_holdout_pred
    df_results['Error'] = df_results[target_column] - df_results['Predicted']
    df_results['Absolute Error'] = df_results['Error'].abs()
    df_results['APE (%)'] = (df_results['Absolute Error'] / df_results[target_column]) * 100
    
    # Calculate R² for each sample (this is needed for group stats)
    df_results['R² Contribution'] = df_results.apply(
        lambda row: (row[target_column] - row['Predicted'])**2, axis=1
    )
    
    for col in group_cols:
        if col not in df_results.columns:
            print(f"❌ Column '{col}' not in holdout data. Skipping.")
            continue
        
        print(f"\n📌 MAPE and R² Results by {col}:")
        
        # Aggregate mean ± std MAPE per group
        group_stats = df_results.groupby(col).agg({
            'APE (%)': ['mean', 'std'],
            'R² Contribution': 'sum',
            target_column: ['mean', 'count'],
            'Predicted': 'mean'
        }).reset_index()
        
        # Flatten column names
        group_stats.columns = [f'{col[0]}_{col[1]}' if col[1] else col[0] for col in group_stats.columns]
        group_stats = group_stats.rename(columns={
            f'{col}_': col,
            'APE (%)_mean': 'MAPE_mean',
            'APE (%)_std': 'MAPE_std',
            'R² Contribution_sum': 'SS_res',
            f'{target_column}_mean': 'y_true_mean',
            f'{target_column}_count': 'n_samples',
            'Predicted_mean': 'y_pred_mean'
        })
        
        # Calculate R² for each group
        def calculate_group_r2(row):
            if row['n_samples'] > 1:
                # Calculate total sum of squares
                group_mask = df_results[col] == row[col]
                y_true_group = df_results.loc[group_mask, target_column]
                y_pred_group = df_results.loc[group_mask, 'Predicted']
                ss_res = ((y_true_group - y_pred_group) ** 2).sum()
                ss_tot = ((y_true_group - y_true_group.mean()) ** 2).sum()
                
                if ss_tot > 0:
                    return 1 - (ss_res / ss_tot)
                else:
                    return np.nan
            else:
                return np.nan
        
        group_stats['R²'] = group_stats.apply(calculate_group_r2, axis=1)
        
        # Sort by MAPE (descending - worst to best)
        group_stats = group_stats.sort_values('MAPE_mean', ascending=False)
        
        # Display statistics
        display_stats = group_stats[[col, 'MAPE_mean', 'MAPE_std', 'R²', 'n_samples']].copy()
        display_stats = display_stats.rename(columns={
            'MAPE_mean': 'Mean MAPE (%)',
            'MAPE_std': 'Std MAPE (%)',
            'R²': 'R² Score',
            'n_samples': 'Samples'
        })
        display(display_stats)
        
        # Plot
        groups = group_stats[col].values
        x = np.arange(len(groups))
        width = 0.6
        
        fig, ax = plt.subplots(figsize=(14, 7))
        bars = ax.bar(x, group_stats['MAPE_mean'], width, 
                      yerr=group_stats['MAPE_std'], capsize=5, color='skyblue', alpha=0.8)
        
        # Create a second y-axis for R² scores
        ax2 = ax.twinx()
        
        # Plot R² scores as line with markers
        r2_line = ax2.plot(x, group_stats['R²'], 'ro-', linewidth=2, markersize=8, 
                          label='R² Score', zorder=5)
        
        # Add value labels for MAPE
        for idx, (bar, mean, std, r2, n_samples) in enumerate(zip(
            bars, group_stats['MAPE_mean'], group_stats['MAPE_std'], 
            group_stats['R²'], group_stats['n_samples']
        )):
            # MAPE value label
            label_y_pos = bar.get_height() + (std if np.isfinite(std) else 0) + 0.5
            ax.text(bar.get_x() + bar.get_width()/2,
                    label_y_pos,
                    f"{mean:.1f}%",
                    ha='center', va='bottom', fontsize=10, rotation=0,
                    bbox=dict(boxstyle="round,pad=0.2", facecolor='white', alpha=0.8))
            
            # R² value label (positioned differently based on value)
            if not np.isnan(r2):
                r2_label_y = r2
                va = 'bottom' if r2 >= 0 else 'top'
                ax2.text(idx,
                        r2_label_y,
                        f"R²={r2:.2f}\n(n={int(n_samples)})",
                        ha='center', va=va, fontsize=9, color='red',
                        bbox=dict(boxstyle="round,pad=0.2", facecolor='white', alpha=0.9))
        
        # Formatting
        ax.set_xticks(x)
        ax.set_xticklabels(groups, rotation=45, ha='right', fontsize=12)
        ax.set_ylabel("MAPE (%)", fontsize=14, color='blue')
        ax.tick_params(axis='y', labelcolor='blue')
        
        ax2.set_ylabel("R² Score", fontsize=14, color='red')
        ax2.tick_params(axis='y', labelcolor='red')
        
        # Set y-axis ranges with some padding
        ax.set_ylim(0, max(group_stats['MAPE_mean']) * 1.3)
        ax2.set_ylim(min(-0.2, min(group_stats['R²'].dropna()) - 0.1), 
                    max(1.1, max(group_stats['R²'].dropna()) + 0.1))
        
        # Add horizontal line at R² = 0 for reference
        ax2.axhline(y=0, color='red', linestyle='--', alpha=0.3)
        
        # Title
        ax.set_title(f"{best_model_name}: Holdout Performance per {col}\n"
                    f"Overall R² = {overall_r2:.3f}", 
                    fontsize=15, fontweight='bold')
        
        ax.grid(axis="y", linestyle="--", alpha=0.7)
        
        # Add legend
        from matplotlib.lines import Line2D
        from matplotlib.patches import Patch
        legend_elements = [
            Patch(facecolor='skyblue', alpha=0.8, label='MAPE (%)'),
            Line2D([0], [0], color='red', marker='o', linestyle='-', 
                  label='R² Score', linewidth=2, markersize=8)
        ]
        ax.legend(handles=legend_elements, loc='upper right', fontsize=12)
        
        plt.tight_layout()
        plt.show()
```

#### OUTLIER REMOVAL

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Global variable to track which dataframe to use ----
df_to_clean = None

# ---- Check if df_remaining exists and update df_to_clean ----
def check_and_update_data_source():
    global df_to_clean
    # Check if df_remaining exists (from the holdout split)
    if 'df_remaining' in globals() and df_remaining is not None:
        print("🔄 Using df_remaining (data after holdout removal)")
        df_to_clean = df_remaining
    else:
        print("⚠️ Using original dfUTS (no holdout split detected)")
        df_to_clean = dfUTS

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfUTS_clean, dfUTS_outliers
    
    # Check which dataframe to use
    check_and_update_data_source()
    
    if df_to_clean is None:
        print("❌ No dataframe available for cleaning")
        return
        
    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    
    # Use df_to_clean instead of dfUTS
    dfUTS_clean, dfUTS_outliers = remover.fit_transform(df_to_clean)

    with output_area:
        clear_output(wait=True)
        
        # Show which dataset is being used
        if 'df_remaining' in globals() and df_remaining is not None:
            print("📊 Working with: df_remaining (data for model building)")
        else:
            print("📊 Working with: dfUTS (original full dataset)")
            
        print(f"Input shape:    {df_to_clean.shape}")
        print(f"Cleaned shape:  {dfUTS_clean.shape}")
        print(f"Outliers shape: {dfUTS_outliers.shape}")
        print(f"Removed {df_to_clean.shape[0] - dfUTS_clean.shape[0]} rows as outliers")

        if show_full_checkbox.value:
            print("\n✅ Cleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_clean.to_html() + '</div>'))
            print("\n🚩 Outliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned DataFrame (first 5 rows):")
            display(HTML(dfUTS_clean.head().to_html()))
            print("\n🚩 Outliers (first 5 rows):")
            display(HTML(dfUTS_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

#### preprocessing pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
### import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score

# Columns to aggregate by
group_cols = ['Chemistry & Alloy Family', 'Microstructure Class', 'Nuclear Application']

if df_holdout is not None and best_pipeline is not None:
    target_column = 'Ultimate Tensile Strength (MPa)'   # <-- updated
    
    # Predict on holdout
    X_holdout = df_holdout.drop(columns=[target_column])
    y_holdout = df_holdout[target_column]
    y_holdout_pred = best_pipeline.predict(X_holdout)
    
    # Calculate overall R² score
    overall_r2 = r2_score(y_holdout, y_holdout_pred)
    print(f"📊 Overall R² Score on Holdout Set: {overall_r2:.4f}")
    print(f"📊 Overall MAE: {np.mean(np.abs(y_holdout - y_holdout_pred)):.2f} MPa")
    print(f"📊 Overall RMSE: {np.sqrt(np.mean((y_holdout - y_holdout_pred)**2)):.2f} MPa")
    print(f"📊 Number of samples in holdout: {len(y_holdout)}\n")
    
    # Combine results
    df_results = df_holdout.copy()
    df_results['Predicted'] = y_holdout_pred
    df_results['Error'] = df_results[target_column] - df_results['Predicted']
    df_results['Absolute Error'] = df_results['Error'].abs()
    df_results['APE (%)'] = (df_results['Absolute Error'] / df_results[target_column]) * 100
    
    # Loop over group columns
    for col in group_cols:
        if col not in df_results.columns:
            print(f"❌ Column '{col}' not in holdout data. Skipping.")
            continue
        
        print(f"\n📌 MAPE and R² Results by {col}:")
        
        # Calculate statistics for each group
        group_stats_list = []
        for group_name in df_results[col].unique():
            group_mask = df_results[col] == group_name
            group_data = df_results[group_mask]
            
            if len(group_data) < 2:
                # Cannot calculate R² with less than 2 samples
                group_stats = {
                    col: group_name,
                    'MAPE_mean': group_data['APE (%)'].mean() if len(group_data) > 0 else np.nan,
                    'MAPE_std': group_data['APE (%)'].std() if len(group_data) > 1 else np.nan,
                    'R²': np.nan,
                    'n_samples': len(group_data),
                    'MAE_mean': group_data['Absolute Error'].mean() if len(group_data) > 0 else np.nan,
                    'RMSE': np.sqrt(np.mean(group_data['Error']**2)) if len(group_data) > 0 else np.nan
                }
            else:
                # Calculate R² for this group
                y_true_group = group_data[target_column]
                y_pred_group = group_data['Predicted']
                r2 = r2_score(y_true_group, y_pred_group)
                
                group_stats = {
                    col: group_name,
                    'MAPE_mean': group_data['APE (%)'].mean(),
                    'MAPE_std': group_data['APE (%)'].std(),
                    'R²': r2,
                    'n_samples': len(group_data),
                    'MAE_mean': group_data['Absolute Error'].mean(),
                    'RMSE': np.sqrt(np.mean(group_data['Error']**2))
                }
            
            group_stats_list.append(group_stats)
        
        # Create DataFrame
        group_stats = pd.DataFrame(group_stats_list)
        
        # Sort by MAPE (descending - worst to best)
        group_stats = group_stats.sort_values('MAPE_mean', ascending=False)
        
        # Display statistics
        display_stats = group_stats[[col, 'MAPE_mean', 'MAPE_std', 'R²', 'n_samples', 'MAE_mean', 'RMSE']].copy()
        display_stats = display_stats.rename(columns={
            'MAPE_mean': 'Mean MAPE (%)',
            'MAPE_std': 'Std MAPE (%)',
            'R²': 'R² Score',
            'n_samples': 'Samples',
            'MAE_mean': 'MAE (MPa)',
            'RMSE': 'RMSE (MPa)'
        })
        display(display_stats)
        
        # Plot
        groups = group_stats[col].values
        x = np.arange(len(groups))
        width = 0.6
        
        fig, ax = plt.subplots(figsize=(14, 8))
        
        # Create bars for MAPE
        bars = ax.bar(x, group_stats['MAPE_mean'], width,
                      yerr=group_stats['MAPE_std'], capsize=5, color='skyblue', alpha=0.8,
                      label='MAPE (%)')
        
        # Create a second y-axis for R² scores
        ax2 = ax.twinx()
        
        # Plot R² scores as line with markers
        valid_r2_mask = ~group_stats['R²'].isna()
        if valid_r2_mask.any():
            r2_line = ax2.plot(x[valid_r2_mask], group_stats.loc[valid_r2_mask, 'R²'], 
                              'ro-', linewidth=2.5, markersize=10, markerfacecolor='red',
                              markeredgecolor='darkred', markeredgewidth=1.5,
                              label='R² Score', zorder=5)
        
        # Add value labels for MAPE
        for idx, (bar, mean, std, r2, n_samples) in enumerate(zip(
            bars, group_stats['MAPE_mean'], group_stats['MAPE_std'], 
            group_stats['R²'], group_stats['n_samples']
        )):
            # MAPE value label
            label_y_pos = bar.get_height() + (std if np.isfinite(std) else 0) + 0.5
            ax.text(bar.get_x() + bar.get_width()/2,
                    label_y_pos,
                    f"{mean:.1f}%",
                    ha='center', va='bottom', fontsize=10, rotation=0,
                    bbox=dict(boxstyle="round,pad=0.2", facecolor='white', alpha=0.8))
            
            # R² value label (if available)
            if not np.isnan(r2):
                r2_text = f"R²={r2:.2f}"
                if n_samples < 5:  # Highlight small sample sizes
                    r2_text += f"\nn={int(n_samples)}"
                
                # Position based on R² value
                va = 'bottom' if r2 >= 0 else 'top'
                offset = 0.05 if r2 >= 0 else -0.1
                
                ax2.text(idx, r2 + offset, r2_text,
                        ha='center', va=va, fontsize=9, color='red',
                        bbox=dict(boxstyle="round,pad=0.2", facecolor='white', alpha=0.9))
        
        # Formatting
        ax.set_xticks(x)
        ax.set_xticklabels(groups, rotation=45, ha='right', fontsize=12)
        ax.set_ylabel("MAPE (%)", fontsize=14, color='blue')
        ax.tick_params(axis='y', labelcolor='blue')
        
        ax2.set_ylabel("R² Score", fontsize=14, color='red')
        ax2.tick_params(axis='y', labelcolor='red')
        
        # Set y-axis ranges with some padding
        max_mape = max(group_stats['MAPE_mean'])
        ax.set_ylim(0, max_mape * 1.3)
        
        # For R² axis, handle NaN values
        valid_r2 = group_stats['R²'].dropna()
        if len(valid_r2) > 0:
            r2_min = min(-0.2, valid_r2.min() - 0.1)
            r2_max = max(1.1, valid_r2.max() + 0.1)
            ax2.set_ylim(r2_min, r2_max)
        
        # Add horizontal line at R² = 0 for reference
        ax2.axhline(y=0, color='red', linestyle='--', alpha=0.3, linewidth=1)
        
        # Add horizontal line at R² = 1 (perfect fit) for reference
        ax2.axhline(y=1, color='green', linestyle='--', alpha=0.3, linewidth=1)
        
        # Title with overall metrics
        ax.set_title(f"{best_model_name}: Holdout Performance for Ultimate Tensile Strength per {col}\n"
                    f"Overall R² = {overall_r2:.3f} | Overall MAPE = {np.mean(df_results['APE (%)']):.1f}%", 
                    fontsize=15, fontweight='bold', pad=20)
        
        ax.grid(axis="y", linestyle="--", alpha=0.7)
        
        # Add legend
        from matplotlib.lines import Line2D
        from matplotlib.patches import Patch
        
        # Create custom legend items
        legend_elements = [
            Patch(facecolor='skyblue', alpha=0.8, label='MAPE (%)'),
            Line2D([0], [0], color='red', marker='o', linestyle='-', 
                  label='R² Score', linewidth=2.5, markersize=8),
            Line2D([0], [0], color='red', linestyle='--', alpha=0.3, label='R² = 0'),
            Line2D([0], [0], color='green', linestyle='--', alpha=0.3, label='R² = 1 (Perfect)')
        ]
        
        ax.legend(handles=legend_elements, loc='upper right', fontsize=11)
        
        # Add sample size information below x-axis
        ax.text(0.02, -0.15, f"Total samples: {len(df_results)} | Groups shown: {len(groups)}",
                transform=ax.transAxes, fontsize=10, verticalalignment='top',
                bbox=dict(boxstyle="round,pad=0.3", facecolor='lightgray', alpha=0.7))
        
        plt.tight_layout()
        plt.show()
        
        # Print summary statistics
        print(f"\n📊 Summary for {col}:")
        print(f"   • Best R²: {group_stats['R²'].max():.3f} ({group_stats.loc[group_stats['R²'].idxmax(), col]})")
        print(f"   • Worst R²: {group_stats['R²'].min():.3f} ({group_stats.loc[group_stats['R²'].idxmin(), col]})")
        print(f"   • Best MAPE: {group_stats['MAPE_mean'].min():.1f}% ({group_stats.loc[group_stats['MAPE_mean'].idxmin(), col]})")
        print(f"   • Worst MAPE: {group_stats['MAPE_mean'].max():.1f}% ({group_stats.loc[group_stats['MAPE_mean'].idxmax(), col]})")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_feature_importances(trained_models_dict, top_n=10):
    if not trained_models_dict:
        print("No trained models found.")
        return

    for name, search in trained_models_dict.items():
        try:
            model = search.best_estimator_.named_steps['model']

            # Get feature names from preprocessor
            feature_names = search.best_estimator_ \
                                .named_steps['preprocessor'] \
                                .get_feature_names_out()

            # Tree-based models
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                sorted_idx = np.argsort(importances)[::-1][:top_n]

            # Linear models
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                sorted_idx = np.argsort(importances)[::-1][:top_n]
            else:
                print(f"⚠️ Feature importance not available for {name}. Skipping...")
                continue

            top_features = np.array(feature_names)[sorted_idx]
            top_importances = importances[sorted_idx]

            # Plot
            plt.figure(figsize=(8, 5))
            sns.barplot(x=top_importances, y=top_features)
            plt.title(f"Top {top_n} Features - {name}")
            plt.xlabel("Importance")
            plt.ylabel("Feature")
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"❌ Skipping {name} due to error: {e}")

# Example usage:
plot_feature_importances(trained_models, top_n=10)

```

```python
import ipywidgets as widgets
from IPython.display import display
from sklearn.feature_selection import SelectKBest, f_regression, mutual_info_regression, chi2, VarianceThreshold

# Feature selection methods
feature_selection_methods = {
    'F Regression': f_regression,
    'Mutual Info Regression': mutual_info_regression,
    'Chi-Squared': chi2,
    'Variance Threshold': 'variance'
}

# Dropdown for method
method_selector = widgets.Dropdown(
    options=list(feature_selection_methods.keys()),
    value='F Regression',
    description='Feature Selection Method:',
    style={'description_width': 'initial'}
)

# Slider for k (number of features)
k_selector = widgets.IntSlider(
    value=5,
    min=1,
    max=50,
    step=1,
    description='Number of Features (k):',
    style={'description_width': 'initial'}
)

display(method_selector, k_selector)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from skopt import BayesSearchCV
import numpy as np

# Store trained models, predictions, and search objects for later use
trained_models = {}
predictions = {}

def run_model_training(X_train, X_test, y_train, y_test, selected_method, selected_k, cv_method='GridSearchCV', n_iter=20):
    global trained_models, predictions
    trained_models.clear()
    predictions.clear()

    for name, (model, params) in models_params.items():
        print(f"🏋️ Training {name} using {cv_method}...")
        pipeline = build_pipeline(selected_method, selected_k, model)
        
        # Select CV method
        if cv_method == 'GridSearchCV':
            search = GridSearchCV(pipeline, params, cv=5, scoring='r2', n_jobs=-1)
        elif cv_method == 'RandomizedSearchCV':
            search = RandomizedSearchCV(pipeline, params, n_iter=n_iter, cv=5, scoring='r2', n_jobs=-1, random_state=42)
        else:  # BayesSearchCV
            search = BayesSearchCV(pipeline, params, n_iter=n_iter, cv=5, scoring='r2', n_jobs=-1, random_state=42)
        
        search.fit(X_train, y_train)
        y_pred = search.predict(X_test)
        
        trained_models[name] = search
        predictions[name] = {'y_test': y_test, 'y_pred': y_pred}

cv_choice = cv_selector.value
run_model_training(X_train, X_test, y_train, y_test, method_selector.value, k_selector.value, cv_method=cv_choice)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest

# ---- OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_contamination=0.05):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_contamination = iso_contamination

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy()
        elif self.method == 'iqr':
            Q1 = df.quantile(0.25)
            Q3 = df.quantile(0.75)
            IQR = Q3 - Q1
            return df[~((df < (Q1 - self.factor * IQR)) | (df > (Q3 + self.factor * IQR))).any(axis=1)]
        elif self.method == 'zscore':
            z = (df - df.mean()) / df.std()
            return df[(np.abs(z) < self.z_thresh).all(axis=1)]
        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_contamination, random_state=42)
            yhat = iso.fit_predict(df)
            return df[yhat == 1]
        else:
            return df.copy()

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Cont.')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfYS_clean
    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_contamination=iso_cont_slider.value
    )
    dfYS_clean = remover.fit_transform(dfYS)

    removed_rows = len(dfYS) - len(dfYS_clean)

    with output_area:
        clear_output(wait=True)
        print(f"Original shape: {dfYS.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Removed rows:   {removed_rows}")
        print("\nCleaned DataFrame (first 10 rows):")
        display(HTML(dfYS_clean.head(10).to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- OutlierRemover Function ----
def remove_outliers(df, method='none', max_outliers=10):
    if method == 'none':
        return df.copy(), df.iloc[0:0]
    
    numeric_df = df.select_dtypes(include=[np.number])
    
    if method == 'iqr':
        Q1 = numeric_df.quantile(0.25)
        Q3 = numeric_df.quantile(0.75)
        IQR = Q3 - Q1
        score = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).sum(axis=1)
    
    elif method == 'zscore':
        z = (numeric_df - numeric_df.mean()) / numeric_df.std()
        score = (np.abs(z) > 3).sum(axis=1)
    
    elif method == 'isoforest':
        iso = IsolationForest(contamination=max_outliers/len(df), random_state=42)
        iso.fit(numeric_df)
        score = pd.Series(-iso.score_samples(numeric_df), index=df.index)
    
    else:
        return df.copy(), df.iloc[0:0]
    
    score_sorted_idx = score.sort_values(ascending=False).index
    outlier_idx = score_sorted_idx[:max_outliers]
    
    dfYS_clean = df.drop(outlier_idx)
    dfYS_outliers = df.loc[outlier_idx]
    
    return dfYS_clean, dfYS_outliers

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:'
)

outlier_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Num Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Update Function ----
def update(_):
    global dfYS_clean, dfYS_outliers  # Make them accessible globally
    with output_area:
        clear_output(wait=True)
        
        dfYS_clean, dfYS_outliers = remove_outliers(dfYS, method_selector.value, outlier_slider.value)
        
        print(f"Original shape: {dfYS.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Outliers shape: {dfYS_outliers.shape}")
        
        if show_full_checkbox.value:
            print("\nCleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_clean.to_html() + '</div>'))
            
            print("\nOutliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_outliers.to_html() + '</div>'))
        else:
            print("\nCleaned DataFrame (first 5 rows):")
            display(HTML(dfYS_clean.head().to_html()))
            
            print("\nOutliers (first 5 rows):")
            display(HTML(dfYS_outliers.head().to_html()))

# ---- Bind Events ----
method_selector.observe(update, names='value')
outlier_slider.observe(update, names='value')
show_full_checkbox.observe(update, names='value')

# ---- Display UI ----
display(widgets.VBox([method_selector, outlier_slider, show_full_checkbox, output_area]))

# Initial run
update(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Evaluate each best model from GridSearch
k = 5
X_full = X_train  # Full cleaned training data
y_full = y_train

cv_results = {}
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 5
X_full = X_train
y_full = y_train

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display the DataFrame (per model, per fold)
import pandas as pd
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model)
summary = df_kfold_results.groupby('Model').agg({
    'MAE': ['mean', 'std'],
    'MSE': ['mean', 'std'],
    'RMSE': ['mean', 'std'],
    'MAPE (%)': ['mean', 'std'],
    'R²': ['mean', 'std']
})
display(summary)
```
