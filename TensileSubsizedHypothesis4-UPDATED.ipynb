4. Scaling
After unit conversion, scale all features together:

StandardScaler (mean=0, std=1) for models like linear regression, neural nets.

MinMaxScaler (0–1) for distance-based models like k-NN.

This prevents large features from dominating due to their magnitude.

5. Handling Missing ppm Values
If an element is below detection limit, ppm may be recorded as blank or zero:

If it’s truly zero, keep it as 0.

If it’s missing due to not being measured, impute (median is safer than mean for skewed data).

Keep in mind: imputation can introduce bias if the missingness is systematic.

6. Consider Feature Engineering
Ratios of ppm elements can be more predictive than absolute amounts for some metallurgy problems (e.g., Cr/Ni ratio).

Grouping: Combine certain ppm elements into “trace impurity index”.

✅ Rule of Thumb:
For machine learning, always bring ppm into the same scale and unit as other composition features before modeling. Otherwise, the algorithm might think that 500 ppm is "huge" compared to 0.5 wt.%, when in reality they’re almost the same magnitude in physical terms.

sequence of data manipulation is very important because you want to avoid data leakage or a situation where data modification affects the integrity of other data in subsequent manipulations.

its best to start from the least complex, gradually manipulate the dataset, and ensure each action was carried out without affecting the integrity of the dataset.

```python
# Handling values such as <40 ppm, 150-400ppm, <0.6 & applying to all chemical composition columns



import pandas as pd
import numpy as np
import re

def parse_chem_value(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # Handle ppm ranges like "150-400ppm"
    range_match = re.match(r'(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        ppm_avg = (low + high) / 2
        return ppm_avg / 10000  # ppm to wt.%
    
    # Handle less-than values e.g. "<40 ppm" or "<0.6"
    lt_match = re.match(r'<\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if lt_match:
        num = float(lt_match.group(1))
        if "ppm" in val_str:
            return num / 10000  # ppm to wt.%
        else:
            return num  # already wt.%
    
    # Handle single ppm values e.g. "200 ppm"
    ppm_match = re.match(r'(\d+(?:\.\d+)?)(?:\s*ppm)', val_str)
    if ppm_match:
        num = float(ppm_match.group(1))
        return num / 10000  # ppm to wt.%
    
    # Handle plain numbers (already wt.%)
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# Columns to modify
cols_to_modify = [
    'C (wt.%)', 'Si (wt.%)', 'Mn (wt.%)', 'P (wt.%)', 'S (wt.%)',
    'Ni (wt.%)', 'Cr (wt.%)', 'Mo (wt.%)', 'Al (wt.%)', 'N (wt.%)',
    'Ti (wt.%)', 'Fe (wt.%)', 'Nb (wt.%)', 'B (wt.%)', 'Cu (wt.%)',
    'V (wt.%)', 'Mg (wt.%)', 'Y (wt.%)', 'Zr (wt.%)', 'Zn (wt.%)',
    'W (wt.%)', 'Sn (wt.%)', 'Co (wt.%)', 'O (wt.,%)', 'Ta (wt.%)',
    'H (wt.%)'
]

# Apply parsing function to each relevant column
for col in cols_to_modify:
    if col in df.columns:
        df[col] = df[col].apply(parse_chem_value)
```

```python
import pandas as pd
import numpy as np

# Condition for Aspect Ratio (L/W): L and W are not NaN and not zero
mask_LW = (
    df['Gauge Length (mm)'].notna() & (df['Gauge Length (mm)'] != 0) &
    df['Width (mm)'].notna() & (df['Width (mm)'] != 0)
)

df.loc[mask_LW, 'Aspect Ratio (L/W)'] = df.loc[mask_LW, 'Gauge Length (mm)'] / df.loc[mask_LW, 'Width (mm)']

# Condition for Aspect Ratio (T/W): T and W are not NaN and not zero
mask_TW = (
    df['Thickness (mm)'].notna() & (df['Thickness (mm)'] != 0) &
    df['Width (mm)'].notna() & (df['Width (mm)'] != 0)
)

df.loc[mask_TW, 'Aspect Ratio (T/W)'] = df.loc[mask_TW, 'Thickness (mm)'] / df.loc[mask_TW, 'Width (mm)']
```

#### Outlier removal

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfYS_clean, dfYS_outliers
    
    # Use df_remaining if it exists, else default to dfYS
    try:
        data_to_process = df_remaining if 'df_remaining' in globals() and df_remaining is not None else dfYS
    except NameError:
        data_to_process = dfYS

    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    dfYS_clean, dfYS_outliers = remover.fit_transform(data_to_process)

    with output_area:
        clear_output(wait=True)
        print(f"Original shape: {data_to_process.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Outliers shape: {dfYS_outliers.shape}")

        if show_full_checkbox.value:
            print("\n✅ Cleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_clean.to_html() + '</div>'))
            print("\n🚩 Outliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned DataFrame (first 5 rows):")
            display(HTML(dfYS_clean.head().to_html()))
            print("\n🚩 Outliers (first 5 rows):")
            display(HTML(dfYS_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Evaluate each best model from GridSearch
k = 10
X_full = X_train  # Full cleaned training data
y_full = y_train

cv_results = {}
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

def evaluate_train_vs_test(best_pipeline, X_train, y_train, X_test, y_test):
    # Predictions
    y_train_pred = best_pipeline.predict(X_train)
    y_test_pred  = best_pipeline.predict(X_test)

    def get_metrics(y_true, y_pred):
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)
        mape = np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100
        r2 = r2_score(y_true, y_pred)
        return {"MSE": mse, "RMSE": rmse, "MAE": mae, "MAPE (%)": mape, "R²": r2}

    train_metrics = get_metrics(y_train, y_train_pred)
    test_metrics  = get_metrics(y_test, y_test_pred)

    return train_metrics, test_metrics
```

```python
def evaluate_with_kfold_materials(X, y, model_pipeline, model_name, material_col, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []

    fold = 1
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Per-sample MAPE
        train_df = X_train.copy()
        train_df["MAPE (%)"] = np.abs((y_train - y_train_pred) / np.clip(y_train, 1e-10, None)) * 100
        val_df = X_val.copy()
        val_df["MAPE (%)"] = np.abs((y_val - y_val_pred) / np.clip(y_val, 1e-10, None)) * 100

        # Aggregate per material
        for mat, group in train_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Train",
                "MAPE": group["MAPE (%)"].mean()
            })
        for mat, group in val_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Validation",
                "MAPE": group["MAPE (%)"].mean()
            })

        fold += 1

    return pd.DataFrame(results)
```

```python
from sklearn.model_selection import KFold
import numpy as np
import pandas as pd

def evaluate_with_kfold_materials(X, y, model_pipeline, model_name, material_col, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []

    fold = 1
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Per-sample MAPE
        train_df = X_train.copy()
        train_df["MAPE (%)"] = np.abs((y_train - y_train_pred) / np.clip(y_train, 1e-10, None)) * 100
        val_df = X_val.copy()
        val_df["MAPE (%)"] = np.abs((y_val - y_val_pred) / np.clip(y_val, 1e-10, None)) * 100

        # Aggregate per material/material group
        for mat, group in train_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Group": mat,   # <-- use "Group" as generic label
                "Dataset": "Train",
                "MAPE": group["MAPE (%)"].mean()
            })
        for mat, group in val_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Group": mat,
                "Dataset": "Validation",
                "MAPE": group["MAPE (%)"].mean()
            })

        fold += 1

    return pd.DataFrame(results)
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Function for evaluation using KFold
def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Use the ENTIRE dataset instead of only X_train and y_train
k = 10
X_full = X   # Full dataset (features)
y_full = y   # Full dataset (target)

cv_results = {}

# Loop through all best models
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    
    # Print results in the same style as your original code
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

%matplotlib inline

def analyze_feature_importance_across_algorithms(trained_models_dict, top_n=10):
    """
    Create interactive dashboard with:
    1. Dropdown to select individual algorithm (shows top/bottom/cumulative features)
    2. Two summary charts showing consensus top/bottom features across ALL algorithms
    """
    
    if not trained_models_dict:
        print("❌ No trained models found!")
        return None
    
    print(f"📊 Analyzing {len(trained_models_dict)} algorithms...")
    
    # ==================== PART 1: EXTRACT FEATURE IMPORTANCES ====================
    all_importances = {}
    feature_importance_data = {}
    
    for model_name, model_info in trained_models_dict.items():
        try:
            # Get model from pipeline
            if hasattr(model_info, 'best_estimator_'):
                pipeline = model_info.best_estimator_
            else:
                pipeline = model_info
            
            # Get actual model
            if hasattr(pipeline, 'named_steps'):
                model = pipeline.named_steps.get('model', pipeline)
            else:
                model = pipeline
            
            # Get feature importances
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                importance_type = "Importance"
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                importance_type = "Coefficient"
            else:
                print(f"⚠️ Skipping {model_name}: No feature importances available")
                continue
            
            importances = np.ravel(importances)
            
            # Get feature names
            feature_names = []
            try:
                if hasattr(pipeline, 'named_steps') and 'preprocessor' in pipeline.named_steps:
                    feature_names = pipeline.named_steps['preprocessor'].get_feature_names_out()
                elif hasattr(pipeline, 'feature_names_in_'):
                    feature_names = pipeline.feature_names_in_
                else:
                    # Try to get from X_train
                    if 'X_train' in locals():
                        feature_names = X_train.columns.tolist()
                    else:
                        feature_names = [f"Feature_{i}" for i in range(len(importances))]
            except:
                feature_names = [f"Feature_{i}" for i in range(len(importances))]
            
            # Clean feature names
            cleaned_names = []
            for name in feature_names:
                if isinstance(name, str) and '__' in name:
                    cleaned_names.append(name.split('__')[-1])
                else:
                    cleaned_names.append(str(name))
            
            # Store data
            feature_importance_data[model_name] = {
                'importances': importances,
                'feature_names': cleaned_names,
                'importance_type': importance_type
            }
            
            # Create DataFrame for this model
            df = pd.DataFrame({
                'Feature': cleaned_names[:len(importances)],
                'Importance': importances,
                'Model': model_name
            })
            
            # Store in all_importances
            all_importances[model_name] = df.sort_values('Importance', ascending=False)
            
            print(f"✅ {model_name}: {len(importances)} features")
            
        except Exception as e:
            print(f"❌ Error processing {model_name}: {str(e)}")
    
    if not all_importances:
        print("❌ No valid feature importance data found!")
        return None
    
    # ==================== PART 2: CREATE CONSENSUS CHARTS ====================
    def create_consensus_charts():
        """Create two charts showing consensus top/bottom features across all algorithms."""
        
        # Combine all data for consensus analysis
        all_features_set = set()
        for df in all_importances.values():
            all_features_set.update(df['Feature'].tolist())
        
        all_features = list(all_features_set)
        
        # Create consensus DataFrame
        consensus_data = []
        for feature in all_features:
            ranks = []
            importances = []
            
            for model_name, df in all_importances.items():
                feature_row = df[df['Feature'] == feature]
                if not feature_row.empty:
                    # Get rank (1 = most important)
                    rank = df[df['Feature'] == feature].index[0] + 1 if not feature_row.empty else len(df)
                    importance = feature_row['Importance'].values[0]
                    
                    ranks.append(rank)
                    importances.append(importance)
            
            if ranks:  # Only include features found in at least one model
                avg_rank = np.mean(ranks)
                avg_importance = np.mean(importances)
                consensus_data.append({
                    'Feature': feature,
                    'Avg_Rank': avg_rank,
                    'Avg_Importance': avg_importance,
                    'Num_Models': len(ranks)
                })
        
        consensus_df = pd.DataFrame(consensus_data)
        
        # Sort by average rank (lower rank = more important)
        consensus_df = consensus_df.sort_values('Avg_Rank')
        
        # Create figure with two subplots
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
        
        # ========== LEFT: CONSENSUS TOP FEATURES ==========
        top_features = consensus_df.head(top_n).copy()
        top_features = top_features.sort_values('Avg_Importance', ascending=True)  # For horizontal bars
        
        y_pos = np.arange(len(top_features))
        colors = plt.cm.Blues(np.linspace(0.4, 0.9, len(top_features)))
        
        bars1 = ax1.barh(y_pos, top_features['Avg_Importance'], height=0.7,
                        color=colors, edgecolor='black', linewidth=0.5)
        
        # Add value labels
        for bar, importance, rank in zip(bars1, top_features['Avg_Importance'], top_features['Avg_Rank']):
            ax1.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.3f}\n(rank: {rank:.1f})',
                    ha='left', va='center', fontsize=9)
        
        ax1.set_yticks(y_pos)
        # Truncate long feature names
        labels = [label[:25] + '...' if len(label) > 25 else label 
                 for label in top_features['Feature']]
        ax1.set_yticklabels(labels, fontsize=9)
        
        ax1.set_xlabel('Average Importance', fontsize=11)
        ax1.set_title(f'Consensus Top {top_n} Features\n(Across {len(all_importances)} Algorithms)', 
                     fontsize=13, fontweight='bold', pad=15)
        ax1.grid(True, alpha=0.3, axis='x')
        
        # ========== RIGHT: CONSENSUS BOTTOM FEATURES ==========
        bottom_features = consensus_df.tail(top_n).copy()
        bottom_features = bottom_features.sort_values('Avg_Importance', ascending=True)
        
        y_pos_b = np.arange(len(bottom_features))
        colors_b = plt.cm.Reds(np.linspace(0.4, 0.9, len(bottom_features)))
        
        bars2 = ax2.barh(y_pos_b, bottom_features['Avg_Importance'], height=0.7,
                        color=colors_b, edgecolor='black', linewidth=0.5)
        
        # Add value labels
        for bar, importance, rank in zip(bars2, bottom_features['Avg_Importance'], bottom_features['Avg_Rank']):
            ax2.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.4f}\n(rank: {rank:.1f})',
                    ha='left', va='center', fontsize=9)
        
        ax2.set_yticks(y_pos_b)
        labels_b = [label[:25] + '...' if len(label) > 25 else label 
                   for label in bottom_features['Feature']]
        ax2.set_yticklabels(labels_b, fontsize=9)
        
        ax2.set_xlabel('Average Importance', fontsize=11)
        ax2.set_title(f'Consensus Bottom {top_n} Features\n(Across {len(all_importances)} Algorithms)', 
                     fontsize=13, fontweight='bold', pad=15)
        ax2.grid(True, alpha=0.3, axis='x')
        
        plt.suptitle('Consensus Feature Importance Across All Algorithms', 
                    fontsize=15, fontweight='bold', y=1.02)
        plt.tight_layout()
        plt.show()
        
        # Print consensus summary
        print("\n" + "="*70)
        print("CONSENSUS FEATURE IMPORTANCE SUMMARY")
        print("="*70)
        print(f"\nTop {min(5, len(top_features))} Consensus Features:")
        for i, row in top_features.head(5).iterrows():
            print(f"  {i+1:2d}. {row['Feature']:30} : {row['Avg_Importance']:.4f} (avg rank: {row['Avg_Rank']:.1f})")
        
        print(f"\nAnalyzed {len(consensus_df)} features across {len(all_importances)} algorithms")
        print("="*70)
        
        return consensus_df
    
    # ==================== PART 3: CREATE INDIVIDUAL ALGORITHM VIEW WITH 3 CHARTS ====================
    def plot_individual_algorithm(model_name):
        """Plot top, bottom, and cumulative features for individual algorithm."""
        
        if model_name not in all_importances:
            print(f"❌ No data for {model_name}")
            return
        
        df = all_importances[model_name]
        
        # Create figure with 3 subplots
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))
        
        # ========== CHART 1: TOP FEATURES ==========
        top_df = df.head(top_n).sort_values('Importance', ascending=True)
        y_pos = np.arange(len(top_df))
        colors_top = plt.cm.Blues(np.linspace(0.4, 0.9, len(top_df)))
        
        bars1 = ax1.barh(y_pos, top_df['Importance'], height=0.7,
                        color=colors_top, edgecolor='black', linewidth=0.5)
        
        for bar, importance in zip(bars1, top_df['Importance']):
            ax1.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.3f}', ha='left', va='center', fontsize=9)
        
        ax1.set_yticks(y_pos)
        labels_top = [label[:20] + '...' if len(label) > 20 else label 
                     for label in top_df['Feature']]
        ax1.set_yticklabels(labels_top, fontsize=9)
        ax1.set_xlabel('Importance', fontsize=11)
        ax1.set_title(f'Top {top_n} Features\n{model_name}', fontsize=12, fontweight='bold')
        ax1.grid(True, alpha=0.3, axis='x')
        
        # ========== CHART 2: BOTTOM FEATURES ==========
        bottom_df = df.tail(top_n).sort_values('Importance', ascending=True)
        y_pos_b = np.arange(len(bottom_df))
        colors_bottom = plt.cm.Reds(np.linspace(0.4, 0.9, len(bottom_df)))
        
        bars2 = ax2.barh(y_pos_b, bottom_df['Importance'], height=0.7,
                        color=colors_bottom, edgecolor='black', linewidth=0.5)
        
        for bar, importance in zip(bars2, bottom_df['Importance']):
            ax2.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.4f}', ha='left', va='center', fontsize=9)
        
        ax2.set_yticks(y_pos_b)
        labels_bottom = [label[:20] + '...' if len(label) > 20 else label 
                        for label in bottom_df['Feature']]
        ax2.set_yticklabels(labels_bottom, fontsize=9)
        ax2.set_xlabel('Importance', fontsize=11)
        ax2.set_title(f'Bottom {top_n} Features\n{model_name}', fontsize=12, fontweight='bold')
        ax2.grid(True, alpha=0.3, axis='x')
        
        # ========== CHART 3: CUMULATIVE IMPORTANCE ==========
        sorted_df = df.sort_values('Importance', ascending=False)
        sorted_df['Cumulative'] = sorted_df['Importance'].cumsum()
        sorted_df['Cumulative_Pct'] = sorted_df['Cumulative'] / sorted_df['Importance'].sum() * 100
        
        # Plot cumulative curve
        ax3.plot(range(1, len(sorted_df) + 1), sorted_df['Cumulative_Pct'],
                'o-', color='darkgreen', linewidth=2, markersize=4,
                markevery=max(1, len(sorted_df) // 20))
        
        # Add threshold markers
        thresholds = [50, 80, 90, 95]
        colors_thresh = ['gray', 'orange', 'red', 'darkred']
        
        for threshold, color in zip(thresholds, colors_thresh):
            idx = np.where(sorted_df['Cumulative_Pct'] >= threshold)[0]
            if len(idx) > 0:
                n_features = idx[0] + 1
                ax3.axvline(n_features, color=color, linestyle=':', alpha=0.7, linewidth=1.5)
                ax3.plot(n_features, threshold, 's', color=color, markersize=7, 
                        markeredgecolor='black', zorder=5)
                
                # Position text
                if threshold >= 80:
                    y_offset = -5
                else:
                    y_offset = 5
                
                ax3.text(n_features + 0.02*len(sorted_df), 
                        threshold + y_offset,
                        f'{n_features} features\n({threshold}%)', 
                        fontsize=9, ha='left', va='center', color=color,
                        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9, pad=0.3))
        
        ax3.set_xlabel('Number of Features (Sorted by Importance)', fontsize=11)
        ax3.set_ylabel('Cumulative Importance (%)', fontsize=11)
        ax3.set_title('Cumulative Feature Importance\n' + model_name, 
                     fontsize=12, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.set_ylim(0, 105)
        ax3.set_xlim(0, min(40, len(sorted_df)) + 5)
        
        plt.suptitle(f'Feature Importance Analysis - {model_name}', 
                    fontsize=14, fontweight='bold', y=1.05)
        plt.tight_layout()
        plt.show()
    
    # ==================== PART 4: CREATE INTERACTIVE DASHBOARD ====================
    print("\n" + "="*70)
    print("CREATING INTERACTIVE DASHBOARD")
    print("="*70)
    
    # Create dropdown
    algorithm_dropdown = widgets.Dropdown(
        options=list(all_importances.keys()),
        description='Select Algorithm:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='300px')
    )
    
    # Create output areas
    consensus_output = widgets.Output()
    individual_output = widgets.Output()
    
    # Function to update individual algorithm plot
    def update_individual_plot(change):
        with individual_output:
            clear_output(wait=True)
            plot_individual_algorithm(change['new'])
    
    # Connect dropdown to update function
    algorithm_dropdown.observe(update_individual_plot, names='value')
    
    # Display consensus charts once
    with consensus_output:
        print("📊 Generating consensus charts across all algorithms...")
        consensus_df = create_consensus_charts()
    
    # Display individual algorithm plot for first model
    with individual_output:
        print(f"📈 Displaying feature importance for: {algorithm_dropdown.value}")
        plot_individual_algorithm(algorithm_dropdown.value)
    
    # Create dashboard layout
    dashboard = widgets.VBox([
        widgets.HTML("<h3>📊 Feature Importance Dashboard</h3>"),
        widgets.HTML("<h4>1. Consensus Across All Algorithms:</h4>"),
        consensus_output,
        widgets.HTML("<hr><h4>2. Individual Algorithm Analysis (3 charts):</h4>"),
        algorithm_dropdown,
        individual_output,
        widgets.HTML("<hr><p><i>Use dropdown to select different algorithms</i></p>")
    ])
    
    display(dashboard)
    
    return all_importances

# ==================== USAGE ====================
if __name__ == "__main__":
    # Simply call this function with your trained_models dictionary
    if 'trained_models' in locals() and trained_models:
        print(f"Found {len(trained_models)} trained models")
        feature_data = analyze_feature_importance_across_algorithms(trained_models, top_n=15)
    else:
        print("⚠️ 'trained_models' dictionary not found!")
        print("\nUsage:")
        print("analyze_feature_importance_across_algorithms(trained_models, top_n=10)")
```

```python
# Use cross-validation with the best model to get predictions
def get_cv_predictions_for_target_materials(best_pipeline, X_full, y_full, target_materials, k=5):
    from sklearn.model_selection import KFold
    import numpy as np
    
    # Initialize KFold
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    
    # Create arrays to store results
    all_indices = []
    all_materials = []
    all_actuals = []
    all_predictions = []
    
    # Perform cross-validation
    for train_idx, val_idx in kf.split(X_full):
        X_train_fold, X_val_fold = X_full.iloc[train_idx], X_full.iloc[val_idx]
        y_train_fold, y_val_fold = y_full.iloc[train_idx], y_full.iloc[val_idx]
        
        # Train and predict
        best_pipeline.fit(X_train_fold, y_train_fold)
        y_pred_fold = best_pipeline.predict(X_val_fold)
        
        # Store results with original indices and materials
        all_indices.extend(X_val_fold.index.tolist())
        all_materials.extend(X_val_fold['Material'].tolist())
        all_actuals.extend(y_val_fold.values)
        all_predictions.extend(y_pred_fold)
    
    # Create results dataframe
    results_df = pd.DataFrame({
        'Index': all_indices,
        'Material': all_materials,
        'Actual Yield Strength (MPa)': all_actuals,
        'Predicted Yield Strength (MPa)': all_predictions
    })
    
    # Filter for target materials only
    target_results = results_df[results_df['Material'].isin(target_materials)].copy()
    
    # Sort by Material and Index
    target_results = target_results.sort_values(['Material', 'Index'])
    
    return target_results

# Generate CV predictions table
cv_target_materials_table = get_cv_predictions_for_target_materials(best_pipeline, X_full, y_full, target_materials)

# Display CV table
print(f"\n{'='*80}")
print(f"CROSS-VALIDATION PREDICTIONS FOR TARGET MATERIALS - {best_model_name}")
print(f"{'='*80}")
print(f"Total CV predictions for target materials: {len(cv_target_materials_table)}")
print(f"Coverage: {len(cv_target_materials_table['Material'].unique())} out of {len(target_materials)} target materials")
display(cv_target_materials_table)

# Save CV results to CSV
cv_target_materials_table.to_csv(f'best_model_{best_model_name.replace(" ", "_").lower()}_target_materials_cv_predictions.csv', index=False)
print(f"\nSaved to: best_model_{best_model_name.replace(' ', '_').lower()}_target_materials_cv_predictions.csv")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- OutlierRemover Function ----
def remove_outliers(df, method='none', max_outliers=10):
    if method == 'none':
        return df.copy(), df.iloc[0:0]
    
    numeric_df = df.select_dtypes(include=[np.number])
    
    if method == 'iqr':
        Q1 = numeric_df.quantile(0.25)
        Q3 = numeric_df.quantile(0.75)
        IQR = Q3 - Q1
        score = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).sum(axis=1)
    
    elif method == 'zscore':
        z = (numeric_df - numeric_df.mean()) / numeric_df.std()
        score = (np.abs(z) > 3).sum(axis=1)
    
    elif method == 'isoforest':
        iso = IsolationForest(contamination=max_outliers/len(df), random_state=42)
        iso.fit(numeric_df)
        score = pd.Series(-iso.score_samples(numeric_df), index=df.index)
    
    else:
        return df.copy(), df.iloc[0:0]
    
    score_sorted_idx = score.sort_values(ascending=False).index
    outlier_idx = score_sorted_idx[:max_outliers]
    
    dfUTS_clean = df.drop(outlier_idx)
    dfUTS_outliers = df.loc[outlier_idx]
    
    return dfUTS_clean, dfUTS_outliers

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:'
)

outlier_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Num Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Update Function ----
def update(_):
    global dfUTS_clean, dfUTS_outliers  # Make them accessible globally
    with output_area:
        clear_output(wait=True)
        
        dfUTS_clean, dfUTS_outliers = remove_outliers(dfUTS, method_selector.value, outlier_slider.value)
        
        print(f"Original shape: {dfUTS.shape}")
        print(f"New shape:      {dfUTS_clean.shape}")
        print(f"Outliers shape: {dfUTS_outliers.shape}")
        
        if show_full_checkbox.value:
            print("\nCleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_clean.to_html() + '</div>'))
            
            print("\nOutliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_outliers.to_html() + '</div>'))
        else:
            print("\nCleaned DataFrame (first 5 rows):")
            display(HTML(dfUTS_clean.head().to_html()))
            
            print("\nOutliers (first 5 rows):")
            display(HTML(dfUTS_outliers.head().to_html()))

# ---- Bind Events ----
method_selector.observe(update, names='value')
outlier_slider.observe(update, names='value')
show_full_checkbox.observe(update, names='value')

# ---- Display UI ----
display(widgets.VBox([method_selector, outlier_slider, show_full_checkbox, output_area]))

# Initial run
update(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

#### Inner K-Fold Cross-Validation

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 9
X_full = X_train  # Or X if you want the entire dataset
y_full = y_train  # Or y if you want the entire dataset

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display all folds
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model), sorted by R² mean descending
summary = (
    df_kfold_results
    .groupby('Model')
    .agg({
        'MAE': ['mean', 'std'],
        'MSE': ['mean', 'std'],
        'RMSE': ['mean', 'std'],
        'MAPE (%)': ['mean', 'std'],
        'R²': ['mean', 'std']
    })
)

# Sort by R² mean (descending)
summary = summary.sort_values(('R²', 'mean'), ascending=False)

display(summary)
```

```python
def evaluate_with_kfold_detailed_trainval(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Training metrics
        train_mse = mean_squared_error(y_train, y_train_pred)
        train_rmse = np.sqrt(train_mse)
        train_mae = mean_absolute_error(y_train, y_train_pred)
        train_mape = mean_absolute_percentage_error(y_train, y_train_pred)
        train_r2 = r2_score(y_train, y_train_pred)

        # Validation metrics
        val_mse = mean_squared_error(y_val, y_val_pred)
        val_rmse = np.sqrt(val_mse)
        val_mae = mean_absolute_error(y_val, y_val_pred)
        val_mape = mean_absolute_percentage_error(y_val, y_val_pred)
        val_r2 = r2_score(y_val, y_val_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'Train_MSE': train_mse,
            'Train_RMSE': train_rmse,
            'Train_MAE': train_mae,
            'Train_MAPE (%)': train_mape,
            'Train_R²': train_r2,
            'Val_MSE': val_mse,
            'Val_RMSE': val_rmse,
            'Val_MAE': val_mae,
            'Val_MAPE (%)': val_mape,
            'Val_R²': val_r2
        })
        fold += 1

    return results
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

def evaluate_train_vs_test(best_pipeline, X_train, y_train, X_test, y_test):
    # Predictions
    y_train_pred = best_pipeline.predict(X_train)
    y_test_pred  = best_pipeline.predict(X_test)

    def get_metrics(y_true, y_pred):
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)
        mape = np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100
        r2 = r2_score(y_true, y_pred)
        return {"MSE": mse, "RMSE": rmse, "MAE": mae, "MAPE (%)": mape, "R²": r2}

    train_metrics = get_metrics(y_train, y_train_pred)
    test_metrics  = get_metrics(y_test, y_test_pred)

    return train_metrics, test_metrics
```

```python
def evaluate_with_kfold_materials(X, y, model_pipeline, model_name, material_col, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []

    fold = 1
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Per-sample MAPE
        train_df = X_train.copy()
        train_df["MAPE (%)"] = np.abs((y_train - y_train_pred) / np.clip(y_train, 1e-10, None)) * 100
        val_df = X_val.copy()
        val_df["MAPE (%)"] = np.abs((y_val - y_val_pred) / np.clip(y_val, 1e-10, None)) * 100

        # Aggregate per material
        for mat, group in train_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Train",
                "MAPE": group["MAPE (%)"].mean()
            })
        for mat, group in val_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Validation",
                "MAPE": group["MAPE (%)"].mean()
            })

        fold += 1

    return pd.DataFrame(results)
```

```python
# Use cross-validation with the best model to get predictions
def get_cv_predictions_for_target_materials(best_pipeline, X_full, y_full, target_materials, k=5):
    from sklearn.model_selection import KFold
    import numpy as np
    
    # Initialize KFold
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    
    # Create arrays to store results
    all_indices = []
    all_materials = []
    all_actuals = []
    all_predictions = []
    
    # Perform cross-validation
    for train_idx, val_idx in kf.split(X_full):
        X_train_fold, X_val_fold = X_full.iloc[train_idx], X_full.iloc[val_idx]
        y_train_fold, y_val_fold = y_full.iloc[train_idx], y_full.iloc[val_idx]
        
        # Train and predict
        best_pipeline.fit(X_train_fold, y_train_fold)
        y_pred_fold = best_pipeline.predict(X_val_fold)
        
        # Store results with original indices and materials
        all_indices.extend(X_val_fold.index.tolist())
        all_materials.extend(X_val_fold['Material'].tolist())
        all_actuals.extend(y_val_fold.values)
        all_predictions.extend(y_pred_fold)
    
    # Create results dataframe
    results_df = pd.DataFrame({
        'Index': all_indices,
        'Material': all_materials,
        'Actual Ultimate Tensile Strength (MPa)': all_actuals,
        'Predicted Ultimate Tensile Strength (MPa)': all_predictions
    })
    
    # Filter for target materials only
    target_results = results_df[results_df['Material'].isin(target_materials)].copy()
    
    # Sort by Material and Index
    target_results = target_results.sort_values(['Material', 'Index'])
    
    return target_results

# Generate CV predictions table
cv_target_materials_table = get_cv_predictions_for_target_materials(best_pipeline, X_full, y_full, target_materials)

# Display CV table
print(f"\n{'='*80}")
print(f"CROSS-VALIDATION PREDICTIONS FOR TARGET MATERIALS - {best_model_name}")
print(f"{'='*80}")
print(f"Total CV predictions for target materials: {len(cv_target_materials_table)}")
print(f"Coverage: {len(cv_target_materials_table['Material'].unique())} out of {len(target_materials)} target materials")
display(cv_target_materials_table)

# Save CV results to CSV
cv_target_materials_table.to_csv(f'best_model_{best_model_name.replace(" ", "_").lower()}_target_materials_cv__UTS_predictions.csv', index=False)
print(f"\nSaved to: best_model_{best_model_name.replace(' ', '_').lower()}_target_materials_cv_UTS_predictions.csv")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

%matplotlib inline

def analyze_feature_importance_across_algorithms(trained_models_dict, top_n=10):
    """
    Create interactive dashboard with:
    1. Dropdown to select individual algorithm (shows top/bottom/cumulative features)
    2. Two summary charts showing consensus top/bottom features across ALL algorithms
    """
    
    if not trained_models_dict:
        print("❌ No trained models found!")
        return None
    
    print(f"📊 Analyzing {len(trained_models_dict)} algorithms...")
    
    # ==================== PART 1: EXTRACT FEATURE IMPORTANCES ====================
    all_importances = {}
    feature_importance_data = {}
    
    for model_name, model_info in trained_models_dict.items():
        try:
            # Get model from pipeline
            if hasattr(model_info, 'best_estimator_'):
                pipeline = model_info.best_estimator_
            else:
                pipeline = model_info
            
            # Get actual model
            if hasattr(pipeline, 'named_steps'):
                model = pipeline.named_steps.get('model', pipeline)
            else:
                model = pipeline
            
            # Get feature importances
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                importance_type = "Importance"
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                importance_type = "Coefficient"
            else:
                print(f"⚠️ Skipping {model_name}: No feature importances available")
                continue
            
            importances = np.ravel(importances)
            
            # Get feature names
            feature_names = []
            try:
                if hasattr(pipeline, 'named_steps') and 'preprocessor' in pipeline.named_steps:
                    feature_names = pipeline.named_steps['preprocessor'].get_feature_names_out()
                elif hasattr(pipeline, 'feature_names_in_'):
                    feature_names = pipeline.feature_names_in_
                else:
                    # Try to get from X_train
                    if 'X_train' in locals():
                        feature_names = X_train.columns.tolist()
                    else:
                        feature_names = [f"Feature_{i}" for i in range(len(importances))]
            except:
                feature_names = [f"Feature_{i}" for i in range(len(importances))]
            
            # Clean feature names
            cleaned_names = []
            for name in feature_names:
                if isinstance(name, str) and '__' in name:
                    cleaned_names.append(name.split('__')[-1])
                else:
                    cleaned_names.append(str(name))
            
            # Store data
            feature_importance_data[model_name] = {
                'importances': importances,
                'feature_names': cleaned_names,
                'importance_type': importance_type
            }
            
            # Create DataFrame for this model
            df = pd.DataFrame({
                'Feature': cleaned_names[:len(importances)],
                'Importance': importances,
                'Model': model_name
            })
            
            # Store in all_importances
            all_importances[model_name] = df.sort_values('Importance', ascending=False)
            
            print(f"✅ {model_name}: {len(importances)} features")
            
        except Exception as e:
            print(f"❌ Error processing {model_name}: {str(e)}")
    
    if not all_importances:
        print("❌ No valid feature importance data found!")
        return None
    
    # ==================== PART 2: CREATE CONSENSUS CHARTS ====================
    def create_consensus_charts():
        """Create two charts showing consensus top/bottom features across all algorithms."""
        
        # Combine all data for consensus analysis
        all_features_set = set()
        for df in all_importances.values():
            all_features_set.update(df['Feature'].tolist())
        
        all_features = list(all_features_set)
        
        # Create consensus DataFrame
        consensus_data = []
        for feature in all_features:
            ranks = []
            importances = []
            
            for model_name, df in all_importances.items():
                feature_row = df[df['Feature'] == feature]
                if not feature_row.empty:
                    # Get rank (1 = most important)
                    rank = df[df['Feature'] == feature].index[0] + 1 if not feature_row.empty else len(df)
                    importance = feature_row['Importance'].values[0]
                    
                    ranks.append(rank)
                    importances.append(importance)
            
            if ranks:  # Only include features found in at least one model
                avg_rank = np.mean(ranks)
                avg_importance = np.mean(importances)
                consensus_data.append({
                    'Feature': feature,
                    'Avg_Rank': avg_rank,
                    'Avg_Importance': avg_importance,
                    'Num_Models': len(ranks)
                })
        
        consensus_df = pd.DataFrame(consensus_data)
        
        # Sort by average rank (lower rank = more important)
        consensus_df = consensus_df.sort_values('Avg_Rank')
        
        # Create figure with two subplots
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
        
        # ========== LEFT: CONSENSUS TOP FEATURES ==========
        top_features = consensus_df.head(top_n).copy()
        top_features = top_features.sort_values('Avg_Importance', ascending=True)  # For horizontal bars
        
        y_pos = np.arange(len(top_features))
        colors = plt.cm.Blues(np.linspace(0.4, 0.9, len(top_features)))
        
        bars1 = ax1.barh(y_pos, top_features['Avg_Importance'], height=0.7,
                        color=colors, edgecolor='black', linewidth=0.5)
        
        # Add value labels
        for bar, importance, rank in zip(bars1, top_features['Avg_Importance'], top_features['Avg_Rank']):
            ax1.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.3f}\n(rank: {rank:.1f})',
                    ha='left', va='center', fontsize=9)
        
        ax1.set_yticks(y_pos)
        # Truncate long feature names
        labels = [label[:25] + '...' if len(label) > 25 else label 
                 for label in top_features['Feature']]
        ax1.set_yticklabels(labels, fontsize=9)
        
        ax1.set_xlabel('Average Importance', fontsize=11)
        ax1.set_title(f'Consensus Top {top_n} Features\n(Across {len(all_importances)} Algorithms)', 
                     fontsize=13, fontweight='bold', pad=15)
        ax1.grid(True, alpha=0.3, axis='x')
        
        # ========== RIGHT: CONSENSUS BOTTOM FEATURES ==========
        bottom_features = consensus_df.tail(top_n).copy()
        bottom_features = bottom_features.sort_values('Avg_Importance', ascending=True)
        
        y_pos_b = np.arange(len(bottom_features))
        colors_b = plt.cm.Reds(np.linspace(0.4, 0.9, len(bottom_features)))
        
        bars2 = ax2.barh(y_pos_b, bottom_features['Avg_Importance'], height=0.7,
                        color=colors_b, edgecolor='black', linewidth=0.5)
        
        # Add value labels
        for bar, importance, rank in zip(bars2, bottom_features['Avg_Importance'], bottom_features['Avg_Rank']):
            ax2.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.4f}\n(rank: {rank:.1f})',
                    ha='left', va='center', fontsize=9)
        
        ax2.set_yticks(y_pos_b)
        labels_b = [label[:25] + '...' if len(label) > 25 else label 
                   for label in bottom_features['Feature']]
        ax2.set_yticklabels(labels_b, fontsize=9)
        
        ax2.set_xlabel('Average Importance', fontsize=11)
        ax2.set_title(f'Consensus Bottom {top_n} Features\n(Across {len(all_importances)} Algorithms)', 
                     fontsize=13, fontweight='bold', pad=15)
        ax2.grid(True, alpha=0.3, axis='x')
        
        plt.suptitle('Consensus Feature Importance Across All Algorithms', 
                    fontsize=15, fontweight='bold', y=1.02)
        plt.tight_layout()
        plt.show()
        
        # Print consensus summary
        print("\n" + "="*70)
        print("CONSENSUS FEATURE IMPORTANCE SUMMARY")
        print("="*70)
        print(f"\nTop {min(5, len(top_features))} Consensus Features:")
        for i, row in top_features.head(5).iterrows():
            print(f"  {i+1:2d}. {row['Feature']:30} : {row['Avg_Importance']:.4f} (avg rank: {row['Avg_Rank']:.1f})")
        
        print(f"\nAnalyzed {len(consensus_df)} features across {len(all_importances)} algorithms")
        print("="*70)
        
        return consensus_df
    
    # ==================== PART 3: CREATE INDIVIDUAL ALGORITHM VIEW WITH 3 CHARTS ====================
    def plot_individual_algorithm(model_name):
        """Plot top, bottom, and cumulative features for individual algorithm."""
        
        if model_name not in all_importances:
            print(f"❌ No data for {model_name}")
            return
        
        df = all_importances[model_name]
        
        # Create figure with 3 subplots
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))
        
        # ========== CHART 1: TOP FEATURES ==========
        top_df = df.head(top_n).sort_values('Importance', ascending=True)
        y_pos = np.arange(len(top_df))
        colors_top = plt.cm.Blues(np.linspace(0.4, 0.9, len(top_df)))
        
        bars1 = ax1.barh(y_pos, top_df['Importance'], height=0.7,
                        color=colors_top, edgecolor='black', linewidth=0.5)
        
        for bar, importance in zip(bars1, top_df['Importance']):
            ax1.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.3f}', ha='left', va='center', fontsize=9)
        
        ax1.set_yticks(y_pos)
        labels_top = [label[:20] + '...' if len(label) > 20 else label 
                     for label in top_df['Feature']]
        ax1.set_yticklabels(labels_top, fontsize=9)
        ax1.set_xlabel('Importance', fontsize=11)
        ax1.set_title(f'Top {top_n} Features\n{model_name}', fontsize=12, fontweight='bold')
        ax1.grid(True, alpha=0.3, axis='x')
        
        # ========== CHART 2: BOTTOM FEATURES ==========
        bottom_df = df.tail(top_n).sort_values('Importance', ascending=True)
        y_pos_b = np.arange(len(bottom_df))
        colors_bottom = plt.cm.Reds(np.linspace(0.4, 0.9, len(bottom_df)))
        
        bars2 = ax2.barh(y_pos_b, bottom_df['Importance'], height=0.7,
                        color=colors_bottom, edgecolor='black', linewidth=0.5)
        
        for bar, importance in zip(bars2, bottom_df['Importance']):
            ax2.text(bar.get_width() * 1.01,
                    bar.get_y() + bar.get_height()/2,
                    f'{importance:.4f}', ha='left', va='center', fontsize=9)
        
        ax2.set_yticks(y_pos_b)
        labels_bottom = [label[:20] + '...' if len(label) > 20 else label 
                        for label in bottom_df['Feature']]
        ax2.set_yticklabels(labels_bottom, fontsize=9)
        ax2.set_xlabel('Importance', fontsize=11)
        ax2.set_title(f'Bottom {top_n} Features\n{model_name}', fontsize=12, fontweight='bold')
        ax2.grid(True, alpha=0.3, axis='x')
        
        # ========== CHART 3: CUMULATIVE IMPORTANCE ==========
        sorted_df = df.sort_values('Importance', ascending=False)
        sorted_df['Cumulative'] = sorted_df['Importance'].cumsum()
        sorted_df['Cumulative_Pct'] = sorted_df['Cumulative'] / sorted_df['Importance'].sum() * 100
        
        # Plot cumulative curve
        ax3.plot(range(1, len(sorted_df) + 1), sorted_df['Cumulative_Pct'],
                'o-', color='darkgreen', linewidth=2, markersize=4,
                markevery=max(1, len(sorted_df) // 20))
        
        # Add threshold markers
        thresholds = [50, 80, 90, 95]
        colors_thresh = ['gray', 'orange', 'red', 'darkred']
        
        for threshold, color in zip(thresholds, colors_thresh):
            idx = np.where(sorted_df['Cumulative_Pct'] >= threshold)[0]
            if len(idx) > 0:
                n_features = idx[0] + 1
                ax3.axvline(n_features, color=color, linestyle=':', alpha=0.7, linewidth=1.5)
                ax3.plot(n_features, threshold, 's', color=color, markersize=7, 
                        markeredgecolor='black', zorder=5)
                
                # Position text
                if threshold >= 80:
                    y_offset = -5
                else:
                    y_offset = 5
                
                ax3.text(n_features + 0.02*len(sorted_df), 
                        threshold + y_offset,
                        f'{n_features} features\n({threshold}%)', 
                        fontsize=9, ha='left', va='center', color=color,
                        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9, pad=0.3))
        
        ax3.set_xlabel('Number of Features (Sorted by Importance)', fontsize=11)
        ax3.set_ylabel('Cumulative Importance (%)', fontsize=11)
        ax3.set_title('Cumulative Feature Importance\n' + model_name, 
                     fontsize=12, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.set_ylim(0, 105)
        ax3.set_xlim(0, min(40, len(sorted_df)) + 5)
        
        plt.suptitle(f'Feature Importance Analysis - {model_name}', 
                    fontsize=14, fontweight='bold', y=1.05)
        plt.tight_layout()
        plt.show()
    
    # ==================== PART 4: CREATE INTERACTIVE DASHBOARD ====================
    print("\n" + "="*70)
    print("CREATING INTERACTIVE DASHBOARD")
    print("="*70)
    
    # Create dropdown
    algorithm_dropdown = widgets.Dropdown(
        options=list(all_importances.keys()),
        description='Select Algorithm:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='300px')
    )
    
    # Create output areas
    consensus_output = widgets.Output()
    individual_output = widgets.Output()
    
    # Function to update individual algorithm plot
    def update_individual_plot(change):
        with individual_output:
            clear_output(wait=True)
            plot_individual_algorithm(change['new'])
    
    # Connect dropdown to update function
    algorithm_dropdown.observe(update_individual_plot, names='value')
    
    # Display consensus charts once
    with consensus_output:
        print("📊 Generating consensus charts across all algorithms...")
        consensus_df = create_consensus_charts()
    
    # Display individual algorithm plot for first model
    with individual_output:
        print(f"📈 Displaying feature importance for: {algorithm_dropdown.value}")
        plot_individual_algorithm(algorithm_dropdown.value)
    
    # Create dashboard layout
    dashboard = widgets.VBox([
        widgets.HTML("<h3>📊 Feature Importance Dashboard</h3>"),
        widgets.HTML("<h4>1. Consensus Across All Algorithms:</h4>"),
        consensus_output,
        widgets.HTML("<hr><h4>2. Individual Algorithm Analysis (3 charts):</h4>"),
        algorithm_dropdown,
        individual_output,
        widgets.HTML("<hr><p><i>Use dropdown to select different algorithms</i></p>")
    ])
    
    display(dashboard)
    
    return all_importances

# ==================== USAGE ====================
if __name__ == "__main__":
    # Simply call this function with your trained_models dictionary
    if 'trained_models' in locals() and trained_models:
        print(f"Found {len(trained_models)} trained models")
        feature_data = analyze_feature_importance_across_algorithms(trained_models, top_n=15)
    else:
        print("⚠️ 'trained_models' dictionary not found!")
        print("\nUsage:")
        print("analyze_feature_importance_across_algorithms(trained_models, top_n=10)")
```

#### Model Performance

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np

# ---- Widgets ----
model_selector = widgets.Dropdown(
    options=list(trained_models.keys()),
    description='Select Model:'
)

param_selector = widgets.Dropdown(
    options=[],  # dynamically populated
    description='Hyperparameter:'
)

plot_type_selector = widgets.RadioButtons(
    options=['Score / Convergence', 'Loss Function'],
    description='Plot Type:'
)

output_area = widgets.Output()

# ---- Update hyperparameter options dynamically ----
def update_param_options(change):
    model_name = model_selector.value
    search = trained_models[model_name]
    
    if hasattr(search, "cv_results_"):
        params = search.cv_results_['params'][0].keys()
        param_selector.options = list(params)
    else:
        param_selector.options = []

model_selector.observe(update_param_options, names='value')

# ---- Plot function ----
def plot_dashboard(_):
    model_name = model_selector.value
    param_name = param_selector.value
    plot_type = plot_type_selector.value
    search = trained_models[model_name]
    
    with output_area:
        clear_output(wait=True)
        
        # --- Get the underlying model from pipeline ---
        if hasattr(search, "best_estimator_"):
            model = search.best_estimator_.named_steps.get('model', search.best_estimator_)
        else:
            model = search
        
        plt.figure(figsize=(8,4))
        
        # --- Loss function plots (for boosting models) ---
        if plot_type == 'Loss Function':
            if hasattr(model, 'train_score_'):  # GradientBoosting / AdaBoost
                plt.plot(model.train_score_, marker='o')
                plt.xlabel('Epoch')
                plt.ylabel('Loss (Deviance)')
                plt.title(f'{model_name} - Training Loss per Epoch')
                plt.grid(True)
                plt.show()
            else:
                print(f"{model_name} does not expose a loss function for plotting.")
        
        # --- Convergence / CV score plots ---
        else:
            if hasattr(search, 'cv_results_'):
                scores = search.cv_results_['mean_test_score']
                if param_name:  # Plot vs hyperparameter
                    param_values = [p[param_name] for p in search.cv_results_['params']]
                    plt.plot(param_values, scores, marker='o')
                    plt.xlabel(param_name)
                    plt.ylabel('Mean CV Score (R²)')
                    plt.title(f'{model_name} - Score vs {param_name}')
                else:  # Plot vs iteration
                    plt.plot(range(1, len(scores)+1), scores, marker='o')
                    plt.xlabel('Iteration')
                    plt.ylabel('Mean CV Score (R²)')
                    plt.title(f'{model_name} - Convergence Plot')
                plt.grid(True)
                plt.show()
            else:
                print(f"No CV results available for {model_name}")

# ---- Bind events ----
model_selector.observe(plot_dashboard, names='value')
param_selector.observe(plot_dashboard, names='value')
plot_type_selector.observe(plot_dashboard, names='value')

# ---- Display UI ----
display(widgets.VBox([model_selector, param_selector, plot_type_selector, output_area]))

# ---- Initial plot ----
update_param_options(None)
plot_dashboard(None)
```

#### Predicted vs Actual

```python
import matplotlib.pyplot as plt
import math
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error

def plot_predicted_vs_actual(predictions):
    n_models = len(predictions)
    cols = 2                          # ← 2 charts per row
    rows = math.ceil(n_models / cols)

    # Robust global limits
    all_vals = np.concatenate([
        np.r_[v["y_test"], v["y_pred"]] for v in predictions.values()
    ])
    lo, hi = np.percentile(all_vals, [1, 99])

    fig, axes = plt.subplots(
        rows, cols,
        figsize=(6 * cols, 5 * rows),
        constrained_layout=True
    )
    axes = axes.flatten()

    for ax, (name, data) in zip(axes, predictions.items()):
        y_test = np.asarray(data["y_test"])
        y_pred = np.asarray(data["y_pred"])

        # Scatter
        ax.scatter(
            y_test,
            y_pred,
            s=20,
            alpha=0.65
        )

        # Perfect prediction line
        ax.plot([lo, hi], [lo, hi], "--", linewidth=1.5)

        # Metrics
        r2 = r2_score(y_test, y_pred)
        rmse = np.sqrt(mean_squared_error(y_test, y_pred))
        mape = np.mean(np.abs((y_test - y_pred) / y_test)) * 100

        ax.text(
            0.04, 0.96,
            f"$R^2$ = {r2:.3f}\nRMSE = {rmse:.3f}\nMAPE = {mape:.2f}%",
            transform=ax.transAxes,
            va="top",
            fontsize=10,
            bbox=dict(
                boxstyle="round",
                facecolor="white",
                alpha=0.9
            )
        )

        ax.set_title(name, fontsize=13)
        ax.set_xlabel("Actual")
        ax.set_ylabel("Predicted")

        ax.set_xlim(lo, hi)
        ax.set_ylim(lo, hi)
        ax.set_aspect("equal", adjustable="box")

        # Grid + closed box look
        ax.grid(True, alpha=0.3)
        for spine in ax.spines.values():
            spine.set_visible(True)

    # Hide unused axes
    for ax in axes[len(predictions):]:
        ax.set_visible(False)

    fig.suptitle(
        "Predicted vs Actual — Model Comparison",
        fontsize=16
    )

    plt.show()
```
