4. Scaling
After unit conversion, scale all features together:

StandardScaler (mean=0, std=1) for models like linear regression, neural nets.

MinMaxScaler (0–1) for distance-based models like k-NN.

This prevents large features from dominating due to their magnitude.

5. Handling Missing ppm Values
If an element is below detection limit, ppm may be recorded as blank or zero:

If it’s truly zero, keep it as 0.

If it’s missing due to not being measured, impute (median is safer than mean for skewed data).

Keep in mind: imputation can introduce bias if the missingness is systematic.

6. Consider Feature Engineering
Ratios of ppm elements can be more predictive than absolute amounts for some metallurgy problems (e.g., Cr/Ni ratio).

Grouping: Combine certain ppm elements into “trace impurity index”.

✅ Rule of Thumb:
For machine learning, always bring ppm into the same scale and unit as other composition features before modeling. Otherwise, the algorithm might think that 500 ppm is "huge" compared to 0.5 wt.%, when in reality they’re almost the same magnitude in physical terms.

sequence of data manipulation is very important because you want to avoid data leakage or a situation where data modification afffects the integrity of other data in subsequent manipulations.

its best to start from the least complex, gradually manipulate the dataset, and ensure each action was carried out without affecting the integrity of the dataset.

```python
# Handling values such as <40 ppm, 150-400ppm, <0.6 & applying to all chemical composition columns



import pandas as pd
import numpy as np
import re

def parse_chem_value(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # Handle ppm ranges like "150-400ppm"
    range_match = re.match(r'(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        ppm_avg = (low + high) / 2
        return ppm_avg / 10000  # ppm to wt.%
    
    # Handle less-than values e.g. "<40 ppm" or "<0.6"
    lt_match = re.match(r'<\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if lt_match:
        num = float(lt_match.group(1))
        if "ppm" in val_str:
            return num / 10000  # ppm to wt.%
        else:
            return num  # already wt.%
    
    # Handle single ppm values e.g. "200 ppm"
    ppm_match = re.match(r'(\d+(?:\.\d+)?)(?:\s*ppm)', val_str)
    if ppm_match:
        num = float(ppm_match.group(1))
        return num / 10000  # ppm to wt.%
    
    # Handle plain numbers (already wt.%)
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# Columns to modify
cols_to_modify = [
    'C (wt.%)', 'Si (wt.%)', 'Mn (wt.%)', 'P (wt.%)', 'S (wt.%)',
    'Ni (wt.%)', 'Cr (wt.%)', 'Mo (wt.%)', 'Al (wt.%)', 'N (wt.%)',
    'Ti (wt.%)', 'Fe (wt.%)', 'Nb (wt.%)', 'B (wt.%)', 'Cu (wt.%)',
    'V (wt.%)', 'Mg (wt.%)', 'Y (wt.%)', 'Zr (wt.%)', 'Zn (wt.%)',
    'W (wt.%)', 'Sn (wt.%)', 'Co (wt.%)', 'O (wt.,%)', 'Ta (wt.%)',
    'H (wt.%)'
]

# Apply parsing function to each relevant column
for col in cols_to_modify:
    if col in df.columns:
        df[col] = df[col].apply(parse_chem_value)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfYS_clean, dfYS_outliers
    
    # Use df_remaining if it exists, else default to dfYS
    try:
        data_to_process = df_remaining if 'df_remaining' in globals() and df_remaining is not None else dfYS
    except NameError:
        data_to_process = dfYS

    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    dfYS_clean, dfYS_outliers = remover.fit_transform(data_to_process)

    with output_area:
        clear_output(wait=True)
        print(f"Original shape: {data_to_process.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Outliers shape: {dfYS_outliers.shape}")

        if show_full_checkbox.value:
            print("\n✅ Cleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_clean.to_html() + '</div>'))
            print("\n🚩 Outliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned DataFrame (first 5 rows):")
            display(HTML(dfYS_clean.head().to_html()))
            print("\n🚩 Outliers (first 5 rows):")
            display(HTML(dfYS_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

#### Preprocessing Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Evaluate each best model from GridSearch
k = 9
X_full = X_train  # Full cleaned training data
y_full = y_train

cv_results = {}
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 10
X_full = X_train
y_full = y_train

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display the DataFrame (per model, per fold)
import pandas as pd
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model)
summary = df_kfold_results.groupby('Model').agg({
    'MAE': ['mean', 'std'],
    'MSE': ['mean', 'std'],
    'RMSE': ['mean', 'std'],
    'MAPE (%)': ['mean', 'std'],
    'R²': ['mean', 'std']
})
display(summary)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global df_remaining_clean, df_remaining_outliers
    
    # Check if df_remaining exists (created from material split)
    if df_remaining is None:
        with output_area:
            clear_output(wait=True)
            print("⚠️ Warning: df_remaining is not defined!")
            print("Please first run the material selection code to split your data.")
            print(f"Current df_remaining: {df_remaining}")
            return
    
    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    
    # Apply outlier removal to df_remaining instead of dfUTS
    df_remaining_clean, df_remaining_outliers = remover.fit_transform(df_remaining)

    with output_area:
        clear_output(wait=True)
        print(f"Original remaining data shape: {df_remaining.shape}")
        print(f"Cleaned remaining data shape:  {df_remaining_clean.shape}")
        print(f"Outliers removed from remaining: {df_remaining_outliers.shape}")
        print(f"Holdout material: {selected_material if selected_material else 'Not selected'}")
        print(f"Holdout data shape: {df_holdout.shape if df_holdout is not None else 'N/A'}")

        if show_full_checkbox.value:
            print("\n✅ Cleaned Remaining DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + df_remaining_clean.to_html() + '</div>'))
            print("\n🚩 Outliers Removed from Remaining Data (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + df_remaining_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned Remaining DataFrame (first 5 rows):")
            display(HTML(df_remaining_clean.head().to_html()))
            print("\n🚩 Outliers Removed from Remaining Data (first 5 rows):")
            display(HTML(df_remaining_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run - only if df_remaining exists
if df_remaining is not None:
    apply_outlier_removal(None)
else:
    with output_area:
        print("⚠️ Please first select a material to split the data.")
        print("Run the material selection code above to create df_remaining.")
```

#### preprocessing pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from skopt import BayesSearchCV
import numpy as np

# Store trained models, predictions, and search objects for later use
trained_models = {}
predictions = {}

def run_model_training(X_train, X_test, y_train, y_test, selected_method, selected_k, cv_method='GridSearchCV', n_iter=20):
    global trained_models, predictions
    trained_models.clear()
    predictions.clear()

    for name, (model, params) in models_params.items():
        print(f"🏋️ Training {name} using {cv_method}...")
        pipeline = build_pipeline(selected_method, selected_k, model)
        
        # Select CV method
        if cv_method == 'GridSearchCV':
            search = GridSearchCV(pipeline, params, cv=5, scoring='r2', n_jobs=-1)
        elif cv_method == 'RandomizedSearchCV':
            search = RandomizedSearchCV(pipeline, params, n_iter=n_iter, cv=5, scoring='r2', n_jobs=-1, random_state=42)
        else:  # BayesSearchCV
            search = BayesSearchCV(pipeline, params, n_iter=n_iter, cv=5, scoring='r2', n_jobs=-1, random_state=42)
        
        search.fit(X_train, y_train)
        y_pred = search.predict(X_test)
        
        trained_models[name] = search
        predictions[name] = {'y_test': y_test, 'y_pred': y_pred}

cv_choice = cv_selector.value
run_model_training(X_train, X_test, y_train, y_test, method_selector.value, k_selector.value, cv_method=cv_choice)
```

```python
def show_top_features(feature_names):
    for name, search in trained_models.items():
        print(f"\n📊 Top features for {name}:")
        model = search.best_estimator_.named_steps['model']
        
        # Tree-based models
        if hasattr(model, 'feature_importances_'):
            importances = model.feature_importances_
            sorted_idx = np.argsort(importances)[::-1]
            for idx in sorted_idx[:10]:  # top 10
                print(f"{feature_names[idx]}: {importances[idx]:.4f}")
        
        # Linear models
        elif hasattr(model, 'coef_'):
            coefs = np.abs(model.coef_)
            sorted_idx = np.argsort(coefs)[::-1]
            for idx in sorted_idx[:10]:
                print(f"{feature_names[idx]}: {coefs[idx]:.4f}")
        else:
            print("⚠️ Feature importance not available for this model.")

# Get feature names from preprocessing step
feature_names = trained_models[list(trained_models.keys())[0]].best_estimator_.named_steps['preprocessor'].get_feature_names_out()
show_top_features(feature_names)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_feature_importances(trained_models_dict, top_n=10):
    if not trained_models_dict:
        print("No trained models found.")
        return

    for name, search in trained_models_dict.items():
        try:
            model = search.best_estimator_.named_steps['model']

            # Get feature names from preprocessor
            feature_names = search.best_estimator_ \
                                .named_steps['preprocessor'] \
                                .get_feature_names_out()

            # Tree-based models
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                sorted_idx = np.argsort(importances)[::-1][:top_n]

            # Linear models
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                sorted_idx = np.argsort(importances)[::-1][:top_n]
            else:
                print(f"⚠️ Feature importance not available for {name}. Skipping...")
                continue

            top_features = np.array(feature_names)[sorted_idx]
            top_importances = importances[sorted_idx]

            # Plot
            plt.figure(figsize=(8, 5))
            sns.barplot(x=top_importances, y=top_features)
            plt.title(f"Top {top_n} Features - {name}")
            plt.xlabel("Importance")
            plt.ylabel("Feature")
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"❌ Skipping {name} due to error: {e}")

# Example usage:
plot_feature_importances(trained_models, top_n=10)

```
