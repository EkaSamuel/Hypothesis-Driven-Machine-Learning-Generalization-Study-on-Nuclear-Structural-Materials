4. Scaling
After unit conversion, scale all features together:

StandardScaler (mean=0, std=1) for models like linear regression, neural nets.

MinMaxScaler (0–1) for distance-based models like k-NN.

This prevents large features from dominating due to their magnitude.

5. Handling Missing ppm Values
If an element is below detection limit, ppm may be recorded as blank or zero:

If it’s truly zero, keep it as 0.

If it’s missing due to not being measured, impute (median is safer than mean for skewed data).

Keep in mind: imputation can introduce bias if the missingness is systematic.

6. Consider Feature Engineering
Ratios of ppm elements can be more predictive than absolute amounts for some metallurgy problems (e.g., Cr/Ni ratio).

Grouping: Combine certain ppm elements into “trace impurity index”.

✅ Rule of Thumb:
For machine learning, always bring ppm into the same scale and unit as other composition features before modeling. Otherwise, the algorithm might think that 500 ppm is "huge" compared to 0.5 wt.%, when in reality they’re almost the same magnitude in physical terms.

sequence of data manipulation is very important because you want to avoid data leakage or a situation where data modification afffects the integrity of other data in subsequent manipulations.

its best to start from the least complex, gradually manipulate the dataset, and ensure each action was carried out without affecting the integrity of the dataset.

```python
# Handling values such as <40 ppm, 150-400ppm, <0.6 & applying to all chemical composition columns



import pandas as pd
import numpy as np
import re

def parse_chem_value(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # Handle ppm ranges like "150-400ppm"
    range_match = re.match(r'(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if range_match:
        low = float(range_match.group(1))
        high = float(range_match.group(2))
        ppm_avg = (low + high) / 2
        return ppm_avg / 10000  # ppm to wt.%
    
    # Handle less-than values e.g. "<40 ppm" or "<0.6"
    lt_match = re.match(r'<\s*(\d+(?:\.\d+)?)(?:\s*ppm)?', val_str)
    if lt_match:
        num = float(lt_match.group(1))
        if "ppm" in val_str:
            return num / 10000  # ppm to wt.%
        else:
            return num  # already wt.%
    
    # Handle single ppm values e.g. "200 ppm"
    ppm_match = re.match(r'(\d+(?:\.\d+)?)(?:\s*ppm)', val_str)
    if ppm_match:
        num = float(ppm_match.group(1))
        return num / 10000  # ppm to wt.%
    
    # Handle plain numbers (already wt.%)
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# Columns to modify
cols_to_modify = [
    'C (wt.%)', 'Si (wt.%)', 'Mn (wt.%)', 'P (wt.%)', 'S (wt.%)',
    'Ni (wt.%)', 'Cr (wt.%)', 'Mo (wt.%)', 'Al (wt.%)', 'N (wt.%)',
    'Ti (wt.%)', 'Fe (wt.%)', 'Nb (wt.%)', 'B (wt.%)', 'Cu (wt.%)',
    'V (wt.%)', 'Mg (wt.%)', 'Y (wt.%)', 'Zr (wt.%)', 'Zn (wt.%)',
    'W (wt.%)', 'Sn (wt.%)', 'Co (wt.%)', 'O (wt.,%)', 'Ta (wt.%)',
    'H (wt.%)'
]

# Apply parsing function to each relevant column
for col in cols_to_modify:
    if col in df.columns:
        df[col] = df[col].apply(parse_chem_value)
```

```python
import pandas as pd
import numpy as np

# Condition for Aspect Ratio (L/W): L and W are not NaN and not zero
mask_LW = (
    df['Gauge Length (mm)'].notna() & (df['Gauge Length (mm)'] != 0) &
    df['Width (mm)'].notna() & (df['Width (mm)'] != 0)
)

df.loc[mask_LW, 'Aspect Ratio (L/W)'] = df.loc[mask_LW, 'Gauge Length (mm)'] / df.loc[mask_LW, 'Width (mm)']

# Condition for Aspect Ratio (T/W): T and W are not NaN and not zero
mask_TW = (
    df['Thickness (mm)'].notna() & (df['Thickness (mm)'] != 0) &
    df['Width (mm)'].notna() & (df['Width (mm)'] != 0)
)

df.loc[mask_TW, 'Aspect Ratio (T/W)'] = df.loc[mask_TW, 'Thickness (mm)'] / df.loc[mask_TW, 'Width (mm)']
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- Combined OutlierRemover Class ----
class OutlierRemover:
    def __init__(self, method='none', factor=1.5, z_thresh=3.0, iso_cont=0.05, max_outliers=None):
        self.method = method
        self.factor = factor
        self.z_thresh = z_thresh
        self.iso_cont = iso_cont
        self.max_outliers = max_outliers

    def fit_transform(self, df):
        if self.method == 'none':
            return df.copy(), df.iloc[0:0]

        numeric_df = df.select_dtypes(include=[np.number])

        if self.method == 'iqr':
            Q1 = numeric_df.quantile(0.25)
            Q3 = numeric_df.quantile(0.75)
            IQR = Q3 - Q1
            score = ((numeric_df < (Q1 - self.factor * IQR)) | (numeric_df > (Q3 + self.factor * IQR))).sum(axis=1)

        elif self.method == 'zscore':
            z = (numeric_df - numeric_df.mean()) / numeric_df.std()
            score = (np.abs(z) > self.z_thresh).sum(axis=1)

        elif self.method == 'isoforest':
            iso = IsolationForest(contamination=self.iso_cont, random_state=42)
            iso.fit(numeric_df)
            score = pd.Series(-iso.score_samples(numeric_df), index=df.index)

        else:
            return df.copy(), df.iloc[0:0]

        # Sort by score
        score_sorted_idx = score.sort_values(ascending=False).index
        if self.max_outliers:
            outlier_idx = score_sorted_idx[:self.max_outliers]
        else:
            outlier_idx = score[score > 0].index  # Keep all flagged rows

        df_clean = df.drop(outlier_idx)
        df_outliers = df.loc[outlier_idx]

        return df_clean, df_outliers


# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Method:',
    style={'description_width': 'initial'}
)

factor_slider = widgets.FloatSlider(value=1.5, min=0.5, max=3.0, step=0.1, description='IQR Factor')
z_thresh_slider = widgets.FloatSlider(value=3.0, min=1.0, max=5.0, step=0.1, description='Z-score Threshold')
iso_cont_slider = widgets.FloatSlider(value=0.05, min=0.01, max=0.5, step=0.01, description='IsoForest Contamination')
max_outliers_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Top N Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Slider Visibility ----
def update_slider_visibility(change):
    method = change['new']
    factor_slider.layout.display = 'none'
    z_thresh_slider.layout.display = 'none'
    iso_cont_slider.layout.display = 'none'
    if method == 'iqr':
        factor_slider.layout.display = 'block'
    elif method == 'zscore':
        z_thresh_slider.layout.display = 'block'
    elif method == 'isoforest':
        iso_cont_slider.layout.display = 'block'

method_selector.observe(update_slider_visibility, names='value')
update_slider_visibility({'new': method_selector.value})  # Initialize

# ---- Apply Outlier Removal ----
def apply_outlier_removal(_):
    global dfYS_clean, dfYS_outliers
    
    # Use df_remaining if it exists, else default to dfYS
    try:
        data_to_process = df_remaining if 'df_remaining' in globals() and df_remaining is not None else dfYS
    except NameError:
        data_to_process = dfYS

    remover = OutlierRemover(
        method=method_selector.value,
        factor=factor_slider.value,
        z_thresh=z_thresh_slider.value,
        iso_cont=iso_cont_slider.value,
        max_outliers=max_outliers_slider.value
    )
    dfYS_clean, dfYS_outliers = remover.fit_transform(data_to_process)

    with output_area:
        clear_output(wait=True)
        print(f"Original shape: {data_to_process.shape}")
        print(f"New shape:      {dfYS_clean.shape}")
        print(f"Outliers shape: {dfYS_outliers.shape}")

        if show_full_checkbox.value:
            print("\n✅ Cleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_clean.to_html() + '</div>'))
            print("\n🚩 Outliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfYS_outliers.to_html() + '</div>'))
        else:
            print("\n✅ Cleaned DataFrame (first 5 rows):")
            display(HTML(dfYS_clean.head().to_html()))
            print("\n🚩 Outliers (first 5 rows):")
            display(HTML(dfYS_outliers.head().to_html()))

# ---- Bind events ----
method_selector.observe(apply_outlier_removal, names='value')
factor_slider.observe(apply_outlier_removal, names='value')
z_thresh_slider.observe(apply_outlier_removal, names='value')
iso_cont_slider.observe(apply_outlier_removal, names='value')
max_outliers_slider.observe(apply_outlier_removal, names='value')
show_full_checkbox.observe(apply_outlier_removal, names='value')

# ---- Display UI ----
ui = widgets.VBox([method_selector, factor_slider, z_thresh_slider, iso_cont_slider,
                   max_outliers_slider, show_full_checkbox, output_area])
display(ui)

# Initial run
apply_outlier_removal(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def evaluate_models_separated(predictions, trained_models):
    metrics_results = {}
    hyperparams_results = {}

    for name, data in predictions.items():
        y_test, y_pred = data['y_test'], data['y_pred']
        
        # Evaluation metrics
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_test, y_pred)
        mape = np.mean(np.abs((y_test - y_pred) / np.clip(y_test, 1e-10, None))) * 100
        r2 = r2_score(y_test, y_pred)
        
        metrics_results[name] = {
            'MSE': mse,
            'RMSE': rmse,
            'MAE': mae,
            'MAPE (%)': mape,
            'R2 Score': r2
        }
        
        # Hyperparameters
        hyperparams_results[name] = trained_models[name].best_params_
    
    # Convert to DataFrames
    metrics_df = pd.DataFrame(metrics_results).T.sort_values(by='R2 Score', ascending=False)
    hyperparams_df = pd.DataFrame(hyperparams_results).T
    
    return metrics_df, hyperparams_df

# Run evaluation
metrics_df, hyperparams_df = evaluate_models_separated(predictions, trained_models)

# Display results
print("=== Evaluation Metrics ===")
display(metrics_df)

print("\n=== Best Hyperparameters ===")
display(hyperparams_df)
```

```python
from sklearn.model_selection import cross_val_score, cross_validate, KFold
from sklearn.metrics import make_scorer, mean_squared_error, mean_absolute_error
import numpy as np

def evaluate_models_cross_validation(trained_models, X_train, y_train, cv=5):
    """
    Perform cross-validation on trained models
    
    Parameters:
    -----------
    trained_models : dict
        Dictionary of trained GridSearchCV objects
    X_train, y_train : arrays
        Training data
    cv : int
        Number of cross-validation folds
    """
    
    # Define scoring metrics with correct scikit-learn strings
    def mape_scorer(y_true, y_pred):
        return np.mean(np.abs((y_true - y_pred) / np.clip(np.abs(y_true), 1e-10, None))) * 100
    
    def rmse_scorer(y_true, y_pred):
        return np.sqrt(mean_squared_error(y_true, y_pred))
    
    scoring = {
        'neg_mse': 'neg_mean_squared_error',
        'neg_rmse': make_scorer(rmse_scorer, greater_is_better=False),
        'neg_mae': 'neg_mean_absolute_error',
        'mape': make_scorer(mape_scorer, greater_is_better=False),
        'r2': 'r2'  # CORRECTED: Changed from 'r2_score' to 'r2'
    }
    
    cv_results = {}
    
    for name, model in trained_models.items():
        print(f"\n📊 Performing {cv}-fold cross-validation for {name}...")
        
        # Get the best estimator from GridSearchCV
        best_estimator = model.best_estimator_
        
        # Perform cross-validation
        cv_scores = cross_validate(best_estimator, X_train, y_train, 
                                   cv=cv, scoring=scoring, return_train_score=False)
        
        # Calculate mean and std for each metric
        cv_summary = {}
        for metric in scoring.keys():
            scores = cv_scores[f'test_{metric}']
            
            # Convert negative scores to positive for MSE, RMSE, MAE, MAPE
            if metric.startswith('neg_') or metric == 'mape':
                scores = -scores
                metric_name = metric[4:] if metric.startswith('neg_') else metric
            else:
                metric_name = metric
            
            cv_summary[f'CV Mean {metric_name.upper()}'] = np.mean(scores)
            cv_summary[f'CV Std {metric_name.upper()}'] = np.std(scores)
        
        cv_results[name] = cv_summary
    
    # Convert to DataFrame
    cv_df = pd.DataFrame(cv_results).T
    
    # Sort by R2 score
    if 'CV Mean R2' in cv_df.columns:
        cv_df = cv_df.sort_values(by='CV Mean R2', ascending=False)
    
    return cv_df

# Perform cross-validation
cv_df = evaluate_models_cross_validation(trained_models, X_train, y_train, cv=5)

print("=== Cross-Validation Results (5-fold) ===")
display(cv_df)
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Evaluate each best model from GridSearch
k = 10
X_full = X_train  # Full cleaned training data
y_full = y_train

cv_results = {}
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
def evaluate_with_kfold_detailed_trainval(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Training metrics
        train_mse = mean_squared_error(y_train, y_train_pred)
        train_rmse = np.sqrt(train_mse)
        train_mae = mean_absolute_error(y_train, y_train_pred)
        train_mape = mean_absolute_percentage_error(y_train, y_train_pred)
        train_r2 = r2_score(y_train, y_train_pred)

        # Validation metrics
        val_mse = mean_squared_error(y_val, y_val_pred)
        val_rmse = np.sqrt(val_mse)
        val_mae = mean_absolute_error(y_val, y_val_pred)
        val_mape = mean_absolute_percentage_error(y_val, y_val_pred)
        val_r2 = r2_score(y_val, y_val_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'Train_MSE': train_mse,
            'Train_RMSE': train_rmse,
            'Train_MAE': train_mae,
            'Train_MAPE (%)': train_mape,
            'Train_R²': train_r2,
            'Val_MSE': val_mse,
            'Val_RMSE': val_rmse,
            'Val_MAE': val_mae,
            'Val_MAPE (%)': val_mape,
            'Val_R²': val_r2
        })
        fold += 1

    return results
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

def evaluate_train_vs_test(best_pipeline, X_train, y_train, X_test, y_test):
    # Predictions
    y_train_pred = best_pipeline.predict(X_train)
    y_test_pred  = best_pipeline.predict(X_test)

    def get_metrics(y_true, y_pred):
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)
        mape = np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100
        r2 = r2_score(y_true, y_pred)
        return {"MSE": mse, "RMSE": rmse, "MAE": mae, "MAPE (%)": mape, "R²": r2}

    train_metrics = get_metrics(y_train, y_train_pred)
    test_metrics  = get_metrics(y_test, y_test_pred)

    return train_metrics, test_metrics
```

```python
def evaluate_with_kfold_materials(X, y, model_pipeline, model_name, material_col, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []

    fold = 1
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Per-sample MAPE
        train_df = X_train.copy()
        train_df["MAPE (%)"] = np.abs((y_train - y_train_pred) / np.clip(y_train, 1e-10, None)) * 100
        val_df = X_val.copy()
        val_df["MAPE (%)"] = np.abs((y_val - y_val_pred) / np.clip(y_val, 1e-10, None)) * 100

        # Aggregate per material
        for mat, group in train_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Train",
                "MAPE": group["MAPE (%)"].mean()
            })
        for mat, group in val_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Validation",
                "MAPE": group["MAPE (%)"].mean()
            })

        fold += 1

    return pd.DataFrame(results)
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Function for evaluation using KFold
def evaluate_with_kfold(X, y, model_pipeline, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    mae_scores, rmse_scores, mse_scores, mape_scores, r2_scores = [], [], [], [], []

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        mae_scores.append(mae)
        mse_scores.append(mse)
        rmse_scores.append(rmse)
        mape_scores.append(mape)
        r2_scores.append(r2)

    return {
        'MAE': (np.mean(mae_scores), np.std(mae_scores)),
        'MSE': (np.mean(mse_scores), np.std(mse_scores)),
        'RMSE': (np.mean(rmse_scores), np.std(rmse_scores)),
        'MAPE': (np.mean(mape_scores), np.std(mape_scores)),
        'R2': (np.mean(r2_scores), np.std(r2_scores))
    }

# ✅ Use the ENTIRE dataset instead of only X_train and y_train
k = 10
X_full = X   # Full dataset (features)
y_full = y   # Full dataset (target)

cv_results = {}

# Loop through all best models
for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    metrics = evaluate_with_kfold(X_full, y_full, best_pipeline, k=k)
    cv_results[name] = metrics
    
    # Print results in the same style as your original code
    print(f"\nModel: {name}")
    print(f"MAE:  {metrics['MAE'][0]:.4f} ± {metrics['MAE'][1]:.4f}")
    print(f"MSE:  {metrics['MSE'][0]:.4f} ± {metrics['MSE'][1]:.4f}")
    print(f"RMSE: {metrics['RMSE'][0]:.4f} ± {metrics['RMSE'][1]:.4f}")
    print(f"MAPE: {metrics['MAPE'][0]:.2f}% ± {metrics['MAPE'][1]:.2f}%")
    print(f"R²:   {metrics['R2'][0]:.4f} ± {metrics['R2'][1]:.4f}")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- OutlierRemover Function ----
def remove_outliers(df, method='none', max_outliers=10):
    if method == 'none':
        return df.copy(), df.iloc[0:0]
    
    numeric_df = df.select_dtypes(include=[np.number])
    
    if method == 'iqr':
        Q1 = numeric_df.quantile(0.25)
        Q3 = numeric_df.quantile(0.75)
        IQR = Q3 - Q1
        score = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).sum(axis=1)
    
    elif method == 'zscore':
        z = (numeric_df - numeric_df.mean()) / numeric_df.std()
        score = (np.abs(z) > 3).sum(axis=1)
    
    elif method == 'isoforest':
        iso = IsolationForest(contamination=max_outliers/len(df), random_state=42)
        iso.fit(numeric_df)
        score = pd.Series(-iso.score_samples(numeric_df), index=df.index)
    
    else:
        return df.copy(), df.iloc[0:0]
    
    score_sorted_idx = score.sort_values(ascending=False).index
    outlier_idx = score_sorted_idx[:max_outliers]
    
    dfUTS_clean = df.drop(outlier_idx)
    dfUTS_outliers = df.loc[outlier_idx]
    
    return dfUTS_clean, dfUTS_outliers

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:'
)

outlier_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Num Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Update Function ----
def update(_):
    global dfUTS_clean, dfUTS_outliers  # Make them accessible globally
    with output_area:
        clear_output(wait=True)
        
        dfUTS_clean, dfUTS_outliers = remove_outliers(dfUTS, method_selector.value, outlier_slider.value)
        
        print(f"Original shape: {dfUTS.shape}")
        print(f"New shape:      {dfUTS_clean.shape}")
        print(f"Outliers shape: {dfUTS_outliers.shape}")
        
        if show_full_checkbox.value:
            print("\nCleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_clean.to_html() + '</div>'))
            
            print("\nOutliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUTS_outliers.to_html() + '</div>'))
        else:
            print("\nCleaned DataFrame (first 5 rows):")
            display(HTML(dfUTS_clean.head().to_html()))
            
            print("\nOutliers (first 5 rows):")
            display(HTML(dfUTS_outliers.head().to_html()))

# ---- Bind Events ----
method_selector.observe(update, names='value')
outlier_slider.observe(update, names='value')
show_full_checkbox.observe(update, names='value')

# ---- Display UI ----
display(widgets.VBox([method_selector, outlier_slider, show_full_checkbox, output_area]))

# Initial run
update(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 3
X_full = X_train  # Or X if you want the entire dataset
y_full = y_train  # Or y if you want the entire dataset

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display all folds
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model), sorted by R² mean descending
summary = (
    df_kfold_results
    .groupby('Model')
    .agg({
        'MAE': ['mean', 'std'],
        'MSE': ['mean', 'std'],
        'RMSE': ['mean', 'std'],
        'MAPE (%)': ['mean', 'std'],
        'R²': ['mean', 'std']
    })
)

# Sort by R² mean (descending)
summary = summary.sort_values(('R²', 'mean'), ascending=False)

display(summary)
```

```python
def evaluate_with_kfold_detailed_trainval(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Training metrics
        train_mse = mean_squared_error(y_train, y_train_pred)
        train_rmse = np.sqrt(train_mse)
        train_mae = mean_absolute_error(y_train, y_train_pred)
        train_mape = mean_absolute_percentage_error(y_train, y_train_pred)
        train_r2 = r2_score(y_train, y_train_pred)

        # Validation metrics
        val_mse = mean_squared_error(y_val, y_val_pred)
        val_rmse = np.sqrt(val_mse)
        val_mae = mean_absolute_error(y_val, y_val_pred)
        val_mape = mean_absolute_percentage_error(y_val, y_val_pred)
        val_r2 = r2_score(y_val, y_val_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'Train_MSE': train_mse,
            'Train_RMSE': train_rmse,
            'Train_MAE': train_mae,
            'Train_MAPE (%)': train_mape,
            'Train_R²': train_r2,
            'Val_MSE': val_mse,
            'Val_RMSE': val_rmse,
            'Val_MAE': val_mae,
            'Val_MAPE (%)': val_mape,
            'Val_R²': val_r2
        })
        fold += 1

    return results
```

```python
def evaluate_with_kfold_materials(X, y, model_pipeline, model_name, material_col, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    results = []

    fold = 1
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)

        # Predictions
        y_train_pred = model_pipeline.predict(X_train)
        y_val_pred   = model_pipeline.predict(X_val)

        # Per-sample MAPE
        train_df = X_train.copy()
        train_df["MAPE (%)"] = np.abs((y_train - y_train_pred) / np.clip(y_train, 1e-10, None)) * 100
        val_df = X_val.copy()
        val_df["MAPE (%)"] = np.abs((y_val - y_val_pred) / np.clip(y_val, 1e-10, None)) * 100

        # Aggregate per material
        for mat, group in train_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Train",
                "MAPE": group["MAPE (%)"].mean()
            })
        for mat, group in val_df.groupby(material_col):
            results.append({
                "Model": model_name,
                "Fold": fold,
                "Material": mat,
                "Dataset": "Validation",
                "MAPE": group["MAPE (%)"].mean()
            })

        fold += 1

    return pd.DataFrame(results)
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 5
X_full = X_train
y_full = y_train

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display the DataFrame (per model, per fold)
import pandas as pd
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model)
summary = df_kfold_results.groupby('Model').agg({
    'MAE': ['mean', 'std'],
    'MSE': ['mean', 'std'],
    'RMSE': ['mean', 'std'],
    'MAPE (%)': ['mean', 'std'],
    'R²': ['mean', 'std']
})
display(summary)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- OutlierRemover Function ----
def remove_outliers(df, method='none', max_outliers=10):
    if method == 'none':
        return df.copy(), df.iloc[0:0]
    
    numeric_df = df.select_dtypes(include=[np.number])
    
    if method == 'iqr':
        Q1 = numeric_df.quantile(0.25)
        Q3 = numeric_df.quantile(0.75)
        IQR = Q3 - Q1
        score = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).sum(axis=1)
    
    elif method == 'zscore':
        z = (numeric_df - numeric_df.mean()) / numeric_df.std()
        score = (np.abs(z) > 3).sum(axis=1)
    
    elif method == 'isoforest':
        iso = IsolationForest(contamination=max_outliers/len(df), random_state=42)
        iso.fit(numeric_df)
        score = pd.Series(-iso.score_samples(numeric_df), index=df.index)
    
    else:
        return df.copy(), df.iloc[0:0]
    
    score_sorted_idx = score.sort_values(ascending=False).index
    outlier_idx = score_sorted_idx[:max_outliers]
    
    dfUE_clean = df.drop(outlier_idx)
    dfUE_outliers = df.loc[outlier_idx]
    
    return dfUE_clean, dfUE_outliers

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:'
)

outlier_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Num Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Update Function ----
def update(_):
    global dfUE_clean, dfUE_outliers  # Make them accessible globally
    with output_area:
        clear_output(wait=True)
        
        dfUE_clean, dfUE_outliers = remove_outliers(dfUE, method_selector.value, outlier_slider.value)
        
        print(f"Original shape: {dfUE.shape}")
        print(f"New shape:      {dfUE_clean.shape}")
        print(f"Outliers shape: {dfUE_outliers.shape}")
        
        if show_full_checkbox.value:
            print("\nCleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUE_clean.to_html() + '</div>'))
            
            print("\nOutliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfUE_outliers.to_html() + '</div>'))
        else:
            print("\nCleaned DataFrame (first 5 rows):")
            display(HTML(dfUE_clean.head().to_html()))
            
            print("\nOutliers (first 5 rows):")
            display(HTML(dfUE_outliers.head().to_html()))

# ---- Bind Events ----
method_selector.observe(update, names='value')
outlier_slider.observe(update, names='value')
show_full_checkbox.observe(update, names='value')

# ---- Display UI ----
display(widgets.VBox([method_selector, outlier_slider, show_full_checkbox, output_area]))

# Initial run
update(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 9
X_full = X_train  # Or X if you want the entire dataset
y_full = y_train  # Or y if you want the entire dataset

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display all folds
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model), sorted by R² mean descending
summary = (
    df_kfold_results
    .groupby('Model')
    .agg({
        'MAE': ['mean', 'std'],
        'MSE': ['mean', 'std'],
        'RMSE': ['mean', 'std'],
        'MAPE (%)': ['mean', 'std'],
        'R²': ['mean', 'std']
    })
)

# Sort by R² mean (descending)
summary = summary.sort_values(('R²', 'mean'), ascending=False)

display(summary)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from sklearn.ensemble import IsolationForest
import numpy as np
import pandas as pd

# ---- OutlierRemover Function ----
def remove_outliers(df, method='none', max_outliers=10):
    if method == 'none':
        return df.copy(), df.iloc[0:0]
    
    numeric_df = df.select_dtypes(include=[np.number])
    
    if method == 'iqr':
        Q1 = numeric_df.quantile(0.25)
        Q3 = numeric_df.quantile(0.75)
        IQR = Q3 - Q1
        score = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).sum(axis=1)
    
    elif method == 'zscore':
        z = (numeric_df - numeric_df.mean()) / numeric_df.std()
        score = (np.abs(z) > 3).sum(axis=1)
    
    elif method == 'isoforest':
        iso = IsolationForest(contamination=max_outliers/len(df), random_state=42)
        iso.fit(numeric_df)
        score = pd.Series(-iso.score_samples(numeric_df), index=df.index)
    
    else:
        return df.copy(), df.iloc[0:0]
    
    score_sorted_idx = score.sort_values(ascending=False).index
    outlier_idx = score_sorted_idx[:max_outliers]
    
    dfTE_clean = df.drop(outlier_idx)
    dfTE_outliers = df.loc[outlier_idx]
    
    return dfTE_clean, dfTE_outliers

# ---- Widgets ----
method_selector = widgets.Dropdown(
    options=['none', 'zscore', 'iqr', 'isoforest'],
    value='none',
    description='Outlier Method:'
)

outlier_slider = widgets.IntSlider(value=10, min=1, max=50, step=1, description='Num Outliers')

show_full_checkbox = widgets.Checkbox(value=False, description='Show Full DataFrame')

output_area = widgets.Output()

# ---- Update Function ----
def update(_):
    global dfTE_clean, dfTE_outliers  # Make them accessible globally
    with output_area:
        clear_output(wait=True)
        
        dfTE_clean, dfTE_outliers = remove_outliers(dfTE, method_selector.value, outlier_slider.value)
        
        print(f"Original shape: {dfTE.shape}")
        print(f"New shape:      {dfTE_clean.shape}")
        print(f"Outliers shape: {dfTE_outliers.shape}")
        
        if show_full_checkbox.value:
            print("\nCleaned DataFrame (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfTE_clean.to_html() + '</div>'))
            
            print("\nOutliers (all rows):")
            display(HTML('<div style="max-height:300px; overflow:auto;">' + dfTE_outliers.to_html() + '</div>'))
        else:
            print("\nCleaned DataFrame (first 5 rows):")
            display(HTML(dfTE_clean.head().to_html()))
            
            print("\nOutliers (first 5 rows):")
            display(HTML(dfTE_outliers.head().to_html()))

# ---- Bind Events ----
method_selector.observe(update, names='value')
outlier_slider.observe(update, names='value')
show_full_checkbox.observe(update, names='value')

# ---- Display UI ----
display(widgets.VBox([method_selector, outlier_slider, show_full_checkbox, output_area]))

# Initial run
update(None)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# Identify numeric/categorical features
numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

# Numeric pipeline
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('normalizer', MinMaxScaler())
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Column transformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Pipeline builder with dynamic feature selection
def build_pipeline(selected_method, k, model):
    from sklearn.feature_selection import SelectKBest, VarianceThreshold
    if selected_method == 'Variance Threshold':
        selector = VarianceThreshold(threshold=0.01)
    else:
        selector = SelectKBest(score_func=feature_selection_methods[selected_method], k=k)
    
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('feature_selection', selector),
        ('model', model)
    ])
    return pipeline
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Custom MAPE function
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-10, None))) * 100

# Evaluate with KFold and return per-fold results
def evaluate_with_kfold_detailed(X, y, model_pipeline, model_name, k=5):
    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    results = []
    fold = 1

    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        model_pipeline.fit(X_train, y_train)
        y_pred = model_pipeline.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        mse = mean_squared_error(y_val, y_pred)
        rmse = np.sqrt(mse)
        mape = mean_absolute_percentage_error(y_val, y_pred)
        r2 = r2_score(y_val, y_pred)

        results.append({
            'Model': model_name,
            'Fold': fold,
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE (%)': mape,
            'R²': r2
        })
        fold += 1

    return results

# ✅ Run evaluation for all trained models
k = 9
X_full = X_train  # Or X if you want the entire dataset
y_full = y_train  # Or y if you want the entire dataset

all_results = []

for name, search_obj in trained_models.items():
    best_pipeline = search_obj.best_estimator_
    fold_results = evaluate_with_kfold_detailed(X_full, y_full, best_pipeline, name, k=k)
    all_results.extend(fold_results)

# Convert to DataFrame for detailed per-fold view
df_kfold_results = pd.DataFrame(all_results)

# Display all folds
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
display(df_kfold_results)

# ✅ Summary (mean per model), sorted by R² mean descending
summary = (
    df_kfold_results
    .groupby('Model')
    .agg({
        'MAE': ['mean', 'std'],
        'MSE': ['mean', 'std'],
        'RMSE': ['mean', 'std'],
        'MAPE (%)': ['mean', 'std'],
        'R²': ['mean', 'std']
    })
)

# Sort by R² mean (descending)
summary = summary.sort_values(('R²', 'mean'), ascending=False)

display(summary)
```

```python
def plot_feature_importances(trained_models_dict, top_n=10):
    if not trained_models_dict:
        print("No trained models found.")
        return

    for name, search in trained_models_dict.items():
        try:
            model = search.best_estimator_.named_steps['model']

            # Get feature names from preprocessor
            feature_names = search.best_estimator_ \
                                .named_steps['preprocessor'] \
                                .get_feature_names_out()

            # Tree-based models
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                sorted_idx = np.argsort(importances)[::-1][:top_n]

            # Linear models
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                sorted_idx = np.argsort(importances)[::-1][:top_n]
            else:
                print(f"⚠️ Feature importance not available for {name}. Skipping...")
                continue

            top_features = np.array(feature_names)[sorted_idx]
            top_importances = importances[sorted_idx]

            # Plot
            plt.figure(figsize=(8, 5))
            sns.barplot(x=top_importances, y=top_features)
            plt.title(f"Top {top_n} Features - {name}")
            plt.xlabel("Importance")
            plt.ylabel("Feature")
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"❌ Skipping {name} due to error: {e}")

# Example usage:
plot_feature_importances(trained_models, top_n=10)
```

```python
def plot_feature_importances(trained_models_dict, top_n=10):
    if not trained_models_dict:
        print("No trained models found.")
        return

    for name, search in trained_models_dict.items():
        try:
            model = search.best_estimator_.named_steps['model']

            # Get feature names from preprocessor
            feature_names = search.best_estimator_ \
                                .named_steps['preprocessor'] \
                                .get_feature_names_out()

            # Tree-based models
            if hasattr(model, 'feature_importances_'):
                importances = model.feature_importances_
                sorted_idx = np.argsort(importances)[::-1][:top_n]

            # Linear models
            elif hasattr(model, 'coef_'):
                importances = np.abs(model.coef_)
                sorted_idx = np.argsort(importances)[::-1][:top_n]
            else:
                print(f"⚠️ Feature importance not available for {name}. Skipping...")
                continue

            top_features = np.array(feature_names)[sorted_idx]
            top_importances = importances[sorted_idx]

            # Plot
            plt.figure(figsize=(8, 5))
            sns.barplot(x=top_importances, y=top_features)
            plt.title(f"Top {top_n} Features - {name}")
            plt.xlabel("Importance")
            plt.ylabel("Feature")
            plt.tight_layout()
            plt.show()

        except Exception as e:
            print(f"❌ Skipping {name} due to error: {e}")

# Example usage:
plot_feature_importances(trained_models, top_n=10)
```
